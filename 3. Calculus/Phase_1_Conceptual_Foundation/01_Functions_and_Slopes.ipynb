## 1. What is a Function?

At its core, a **function** is a simple but powerful concept: it's a **rule** that takes an **input**, does something to it, and produces a **single, predictable output**.

Think of it like a **machine**:

- You put in an **ingredient** (the input, $ x $).
- The machine follows a **specific recipe** (the function, $ f $).
- It gives you a **finished product** (the output, $ y $).

### Key Principles

| Term | Meaning |
|------|--------|
| **Notation** | $ f(x) = y $ → "f of x equals y" |
| **Predictability** | Same $ x $ → **always same** $ y $ |
| **Domain** | All valid inputs $ x $ (e.g., $ x ≠ 0 $ for $ 1/x $) |
| **Range** | All possible outputs $ y $ |

### Common Examples

1. **Linear Function**: $ f(x) = mx + b $
   ```python
   f(x) = 2x + 1
   f(3) = 2(3) + 1 = 7
   ```
   Graph: **straight line**

2. **Quadratic Function**: $ f(x) = ax^2 + bx + c $
   ```python
   f(x) = x²
   f(3) = 9, f(-3) = 9  # Different inputs → same output OK
   ```
   Graph: **parabola** (U-shape)

3. **Exponential Function**: $ f(x) = a^x $
   ```python
   f(x) = eˣ  # e ≈ 2.718
   ```
   Used in **softmax**, **attention**, **growth models**

**AI Connection**: A neural network is a **giant function**:
$$ \hat{y} = f_\theta(x) $$
where $ \theta $ = millions of weights we must learn.

````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider
%matplotlib inline
plt.style.use('seaborn-v0_8')

def plot_function_examples():
    x = np.linspace(-3, 3, 400)
    
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.plot(x, 2*x + 1, 'b-', label='f(x) = 2x + 1')
    plt.title('Linear')
    plt.legend()
    
    plt.subplot(1, 3, 2)
    plt.plot(x, x**2, 'g-', label='f(x) = x²')
    plt.title('Quadratic')
    plt.legend()
    
    plt.subplot(1, 3, 3)
    plt.plot(x, np.exp(x), 'r-', label='f(x) = eˣ')
    plt.title('Exponential')
    plt.legend()
    
    plt.tight_layout()
    plt.show()

plot_function_examples()
````

## 2. What is Slope?

The **slope** is a single number that describes the **steepness and direction** of a line.

> **Question**: How much does $ y $ change when $ x $ increases by 1?

### Slope Formula (for a Line)

Given two points $ (x_1, y_1) $ and $ (x_2, y_2) $:

$$ \boxed{m = \frac{\text{Rise}}{\text{Run}} = \frac{y_2 - y_1}{x_2 - x_1} = \frac{\Delta y}{\Delta x}} $$

### Interpreting Slope

| Slope Value | Meaning |
|-------------|--------|
| $ m > 0 $ | Uphill (increasing) |
| $ m < 0 $ | Downhill (decreasing) |
| $ m = 0 $ | Horizontal (constant) |
| Undefined | Vertical (not a function) |

---

**Example**: $ f(x) = x^2 $

Average rate from $ x=1 $ to $ x=3 $:

$$ m = \frac{f(3) - f(1)}{3-1} = \frac{9 - 1}{2} = 4 $$

````python
def average_rate(f, a, b):
    return (f(b) - f(a)) / (b - a)

f = lambda x: x**2
print(f"Average slope from x=1 to x=3: {average_rate(f, 1, 3):.2f}")
````

## 3. The Problem: Slope on a Curve

For **straight lines**, slope is **constant**.

For **curves** like $ f(x) = x^2 $, slope **changes**:

- Steep on sides
- Flat at bottom (vertex)

**How do we find slope at a single point?**

Try plugging in same point:

$$ m = \frac{f(x) - f(x)}{x - x} = \frac{0}{0} $$

**Undefined!** We have a problem.

---

### The Solution: Secant Line

We **approximate** slope using **two close points**:

- Point 1: $ (x, f(x)) $
- Point 2: $ (x+h, f(x+h)) $

Slope of **secant line**:

$$ \boxed{m_{\text{secant}} = \frac{f(x+h) - f(x)}{h}} $$

### The Tangent Line

We want the line that **just touches** the curve at $ x $.

**Idea**: What if we make $ h $ **smaller and smaller**?

→ Secant becomes **tangent**

→ Slope becomes **instantaneous rate of change**

→ This limit is the **derivative** — the heart of calculus.

**AI Link**: In gradient descent, we follow this slope **downhill** on the loss surface.

````python
def plot_secant(x0, h):
    f = lambda x: x**2 
    x = np.linspace(x0 - 2, x0 + 2, 400)
    y = f(x)
    
    plt.figure(figsize=(8, 5))
    plt.plot(x, y, 'b-', label='f(x) = x²')
    
    x1 = x0 + h
    plt.plot([x0, x1], [f(x0), f(x1)], 'r--', linewidth=2, label=f'Secant (h={h})')
    plt.scatter([x0, x1], [f(x0), f(x1)], c='red', zorder=5)
    
    slope = (f(x1) - f(x0)) / h
    plt.title(f'Secant Slope = {slope:.3f} | h = {h}')
    plt.xlabel('x'); plt.ylabel('f(x)')
    plt.legend()
    plt.grid(True)
    plt.show()

interactive_plot = interactive(
    plot_secant,
    x0=FloatSlider(value=1, min=-2, max=2, step=0.1, description='x₀'),
    h=FloatSlider(value=1, min=0.01, max=2, step=0.01, description='h')
)

interactive_plot
````

**Observe**: As $ h \to 0 $, slope → $ 2x $ (we'll prove this with limits!)
