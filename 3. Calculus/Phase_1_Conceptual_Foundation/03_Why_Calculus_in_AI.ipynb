## 1. The Core Problem: How AI Learns

AI (supervised ML) = **optimization problem**:

1. **Model**: $ \hat{y} = f_\theta(x) $ (neural net with parameters $ \theta $)
2. **Loss**: $ L = (\hat{y} - y)^2 $ (how wrong is prediction?)
3. **Goal**: Find $ \theta^* $ that minimizes $ L $

### Example: Mean Squared Error

$$ L = (\hat{y} - y)^2 $$

- Perfect: $ L = 0 $
- Bad: $ L = 16 $

---

## 2. Loss Surface: Finding the Bottom

Imagine $ L(\theta) $ as a **landscape**:

- High error → high up
- Low error → bottom of bowl

**Question**: Which way is **downhill**?

**Answer**: **Calculus** → the **gradient**

---

## 3. The Engine: Gradient Descent

1. Start at random $ \theta $
2. Compute **slope** $ \nabla L $
3. Step **opposite** direction

$$ \boxed{\theta_{\text{new}} = \theta_{\text{old}} - \alpha \cdot \nabla L} $$

- $ \alpha $ = learning rate
- $ \nabla L $ = **calculus derivative**

Repeat → roll downhill → converge to minimum

```python
def gradient_descent_demo(w_start=-2.8, alpha=0.05, steps=25):
    w = w_start
    history = [w]
    
    for _ in range(steps):
        # L(w) = w^4 - 4w^2 + 2w
        # dL/dw = 4w^3 - 8w + 2
        grad = 4*w**3 - 8*w + 2
        
        # The Gradient Descent step
        w = w - alpha * grad
        
        # Safety check to prevent numbers from getting too big
        if abs(w) > 1e4:
            history.append(w)
            break # Stop if it's exploding
        
        history.append(w)
    
    # Plot the loss function (the "bowl")
    w_grid = np.linspace(-3, 3, 400)
    L = w_grid**4 - 4*w_grid**2 + 2*w_grid
    
    plt.figure(figsize=(10, 6))
    plt.plot(w_grid, L, 'b-', linewidth=2, label='Loss Function $L(w)$')
    
    # Plot the path of gradient descent
    history = np.array(history)
    
    L_history = history**4 - 4*history**2 + 2*history
    
    plt.plot(history, L_history, 'ro-', label='GD Path')
    plt.plot(history[0], L_history[0], 'go', markersize=10, label='Start')
    plt.plot(history[-1], L_history[-1], 'rx', markersize=10, label='End')
    
    plt.title(f'Gradient Descent (α = {alpha:.3f}, Start w = {w_start:.1f})')
    plt.xlabel('Weight $w$')
    plt.ylabel('Loss $L$')
    
    # Clip y-axis to prevent plot exploding
    plt.ylim(-10, 50)
    plt.legend()
    plt.grid(True)
    plt.show()

interactive_plot = interactive(gradient_descent_demo,
    w_start=FloatSlider(value=-2.8, min=-3.0, max=3.0, step=0.1, description='Start $w$:'),
    alpha=FloatSlider(value=0.05, min=0.001, max=0.09, step=0.001, description='Learn Rate α:', readout_format='.3f'),
    steps=IntSlider(value=25, min=1, max=100, step=1, description='Steps:')
)

display(interactive_plot)
```

## 4. Deep Networks → Chain Rule → Backpropagation

Neural net: nested functions

$$ L = \text{loss}(\text{out}(\text{layer}_n(\dots \text{layer}_1(x))\dots))) $$

Need: $ \frac{\partial L}{\partial w_1} $

**Chain Rule**:

$$ \frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial \text{out}} \cdot \frac{\partial \text{out}}{\partial \text{layer}_n} \cdots \frac{\partial \text{layer}_1}{\partial w_1} $$

→ **Backpropagation** = **Chain Rule in code**

```python
# PyTorch AutoGrad Demo
x = torch.tensor(1.0)
w1 = torch.tensor(2.0, requires_grad=True)
w2 = torch.tensor(3.0, requires_grad=True)

y = w1 * x + w2 * x**2
target = torch.tensor(10.0)
loss = (y - target)**2

loss.backward()

print(f"∂L/∂w1 = {w1.grad.item():.2f}")
print(f"∂L/∂w2 = {w2.grad.item():.2f}")
```
