## 1. What is a Limit?

A **limit** is the value that $ f(x) $ **approaches** as $ x $ gets **arbitrarily close** to $ c $.

$$ \lim_{x \to c} f(x) = L $$

**Crucial**: $ f(c) $ can be undefined — limit only cares about **behavior near** $ c $.

### Intuitive Example: The Hole

Consider:

$$ f(x) = \frac{x^2 - 1}{x - 1} $$

At $ x = 1 $:

$$ f(1) = \frac{0}{0} \text{ → undefined} $$

But simplify (for $ x \neq 1 $):

$$ f(x) = \frac{(x-1)(x+1)}{x-1} = x + 1 $$

So $ f(x) $ is just the line $ y = x + 1 $ with a **hole** at $ x = 1 $.

**What does it approach?**

- $ x \to 1^- $: $ 0.999 → 1.999 $
- $ x \to 1^+ $: $ 1.001 → 2.001 $

→ Both sides → **2**

$$ \boxed{\lim_{x \to 1} \frac{x^2 - 1}{x - 1} = 2} $$

```python
x = sp.symbols('x')
f = (x**2 - 1)/(x - 1)
limit_val = sp.limit(f, x, 1)

display(Math(r'\lim_{{x\to1}} ' + sp.latex(f) + f' = {limit_val}'))
```

## 2. Connecting Limits to Slope (The Derivative)

From secant slope:

$$ m_{\text{secant}} = \frac{f(x+h) - f(x)}{h} $$

The **tangent slope** at $ x $ is the **limit** as $ h \to 0 $:

$$ \boxed{f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}} $$

This is the **formal definition of the derivative**.

```python
h = sp.symbols('h')
f = x**2
diff_quotient = ((x + h)**2 - x**2) / h
derivative = sp.limit(diff_quotient, h, 0)

display(Math(fr"f'(x) = \lim_{{h\to0}} {sp.latex(diff_quotient)} = {derivative}"))
```

## 3. What is Continuity?

**Intuitive**: You can draw the graph **without lifting your pencil**.

**Formal**: $ f $ is continuous at $ c $ if **all 3 conditions** hold:

1. $ f(c) $ is **defined**
2. $ \lim_{x \to c} f(x) $ **exists**
3. $ \lim_{x \to c} f(x) = f(c) $

### Types of Discontinuity

| Type | Example | Limit Exists? | Fixable? |
|------|--------|---------------|---------|
| **Hole** | $ \frac{x^2-1}{x-1} $ at 1 | Yes | Yes (redefine) |
| **Jump** | Step function | No | No |
| **Asymptote** | $ 1/x $ at 0 | No | No |

**Theorem**: Differentiability → Continuity

---

## Why Continuity Matters in AI

Gradient descent requires a **defined slope**.

- No slope at **jumps** or **holes**
- ReLU has a **corner** at 0 → use **subgradient**
- Loss functions must be **continuous and differentiable**

> **Broken loss = broken training**
