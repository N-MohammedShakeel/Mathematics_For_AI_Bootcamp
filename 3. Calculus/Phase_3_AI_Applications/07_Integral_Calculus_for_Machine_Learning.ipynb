## 1. What is an Integral in Machine Learning?

While **differentiation** is about measuring the instantaneous rate of change (slopes), **integration** is about **accumulation**—calculating total accumulated mass, area, volume, or probabilities across continuous spaces.

### Fundamental Theorem of Calculus:
$$\int_a^b f(x) dx = F(b) - F(a) \quad \text{where } F'(x) = f(x)$$

### Why AI/ML Needs Integral Calculus:
1. **Continuous Random Variables**: In the real world, variables like house prices, voice audio signals, image pixel intensities, and latent vectors are continuous. Probabilities are defined as integrals over Probability Density Functions (PDFs).
2. **Mathematical Expectations**: $\mathbb{E}[g(X)] = \int_{-\infty}^\infty g(x) p(x) dx$.
3. **Normalizing Constants**: Ensuring probabilities integrate to 1: $\int p(x) dx = 1$.
4. **Continuous Generative Models**: Continuous Normalizing Flows and Score-Based Diffusion Models (solving reverse SDEs/ODEs).


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
from scipy import stats

print("SciPy and NumPy loaded successfully!")
```

## 2. Probability Density Functions (PDF) & Cumulative Distributions (CDF)

For a continuous random variable $X$, the probability of landing at a single exact point is 0 ($P(X = 3.14159\dots) = 0$). Instead, probability is the **area under the PDF curve $p(x)$**:

$$P(a \leq X \leq b) = \int_a^b p(x) dx$$

### The Cumulative Distribution Function (CDF):
$$F(x) = P(X \leq x) = \int_{-\infty}^x p(t) dt \quad \iff \quad p(x) = \frac{d}{dx} F(x)$$

### The Normal (Gaussian) Distribution:
$$p(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$


```python
# Computing Gaussian probability via integration
mu, sigma = 0.0, 1.0

# Compute probability P(-1 <= X <= 1) (The 68% Empirical Rule)
pdf_func = lambda x: (1.0 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu)/sigma)**2)
prob_area, err = integrate.quad(pdf_func, -1.0, 1.0)

print(f"Calculated Probability P(-1 <= X <= 1): {prob_area:.4f} (Error estimate: {err:.2e})")

# Visualizing Area Under the Curve
x_vals = np.linspace(-3.5, 3.5, 300)
y_vals = pdf_func(x_vals)

plt.figure(figsize=(9, 5))
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='Standard Normal PDF')
plt.fill_between(x_vals, y_vals, where=((x_vals >= -1.0) & (x_vals <= 1.0)), 
                 color='cyan', alpha=0.5, label=f'Area = {prob_area:.2%} (P[-1, 1])')
plt.title(r"Probability as an Integral: $P(a \leq X \leq b) = \int_a^b p(x)dx$")
plt.xlabel("x"); plt.ylabel("Density p(x)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Continuous Expected Values & Moments

In discrete math, expected value is a summation: $\mathbb{E}[X] = \sum x_i p_i$.
In continuous math, summation becomes an **integral**:

$$\mathbb{E}[X] = \int_{-\infty}^\infty x \cdot p(x) dx$$
$$\text{Var}(X) = \mathbb{E}[(X - \mu)^2] = \int_{-\infty}^\infty (x - \mu)^2 \cdot p(x) dx$$
$$\mathbb{E}[g(X)] = \int_{-\infty}^\infty g(x) \cdot p(x) dx$$


```python
# Compute expectation E[X^2] for Standard Normal (should be 1.0)
exp_x2, _ = integrate.quad(lambda x: (x**2) * pdf_func(x), -np.inf, np.inf)
print(f"E[X^2] for Standard Normal: {exp_x2:.6f}")
```

## 4. Double Integrals & Marginalization

When working with two continuous variables $(X, Y)$ with joint distribution $p(x, y)$:
1. **Total Probability**: $\int_{-\infty}^\infty \int_{-\infty}^\infty p(x, y) dx dy = 1$
2. **Marginal Distribution** (Integrating out the nuisance variable):
$$p(x) = \int_{-\infty}^\infty p(x, y) dy$$


```python
# Numerical Marginalization with 2D Gaussian
def joint_pdf(x, y):
    return (1.0 / (2 * np.pi)) * np.exp(-0.5 * (x**2 + y**2))

# Marginalize out y at x = 0.5 (should equal 1D Gaussian at 0.5)
x_val = 0.5
marginal_x, _ = integrate.quad(lambda y: joint_pdf(x_val, y), -np.inf, np.inf)
true_1d = pdf_func(x_val)

print(f"Marginal p(x={x_val}) via integration: {marginal_x:.6f}")
print(f"True 1D Gaussian p(x={x_val}):        {true_1d:.6f}")
assert np.isclose(marginal_x, true_1d)
print("Marginalization verified!")
```

## 5. Summary & Key Takeaways

1. **Integrals** compute accumulated probability mass, continuous expected values, and normalizing constants.
2. In continuous machine learning, probabilities are **areas under density curves**, defined by $P(a \leq X \leq b) = \int_a^b p(x) dx$.
3. **Marginalization** $p(x) = \int p(x,y) dy$ integrates out latent or unobserved variables.
4. When high-dimensional integrals become intractable to compute analytically, we use **Monte Carlo numerical integration**.
