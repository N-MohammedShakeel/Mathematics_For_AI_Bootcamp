## The Problem: Functions Inside Functions

Imagine you have:

$$ h(x) = \sin(x^{10}) $$

This is **not** $ \sin(x) \cdot x^{10} $ — that would be Product Rule.

This is **$ \sin $ of $ x^{10} $** — a **function inside another function**.

- **Outside function**: $ f(u) = \sin(u) $
- **Inside function**: $ g(x) = x^{10} $

We write: $ h(x) = f(g(x)) $

Another example:

$$ h(x) = (3x^2 + 5)^{100} $$

Expanding this is **impossible**. We need a **smart rule**.

---

## The Chain Rule — The Formula

> If $ h(x) = f(g(x)) $, then:
> 
> $$ \boxed{h'(x) = f'(g(x)) \cdot g'(x)} $$

### The Mantra

> **“Derivative of the outside, leave inside alone… times… derivative of the inside.”**

### Leibniz Notation (AI Gold)

Let $ y = f(u) $, $ u = g(x) $

$$ \frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx} $$

This is **backpropagation in math**.

---

## Example 1: $ h(x) = (x^2 + 3x)^{50} $

- **Outside**: $ f(u) = u^{50} $ → $ f'(u) = 50u^{49} $
- **Inside**: $ g(x) = x^2 + 3x $ → $ g'(x) = 2x + 3 $

Apply mantra:

1. Derivative of outside: $ 50u^{49} $
2. Leave inside: $ 50(x^2 + 3x)^{49} $
3. Times inside derivative: $ \cdot (2x + 3) $

$$ \boxed{h'(x) = 50(x^2 + 3x)^{49} (2x + 3)} $$

---

## Example 2: $ h(x) = e^{5x^2} $

- **Outside**: $ f(u) = e^u $ → $ f'(u) = e^u $
- **Inside**: $ g(x) = 5x^2 $ → $ g'(x) = 10x $

$$ h'(x) = e^{5x^2} \cdot 10x $$

$$ \boxed{= 10x e^{5x^2}} $$

---

## Multi-Layer Chain: $ h(x) = \cos^3(7x) $

Rewrite: $ h(x) = [\cos(7x)]^3 $

Three layers:

1. $ f(u) = u^3 $ → $ f' = 3u^2 $
2. $ g(v) = \cos(v) $ → $ g' = -\sin(v) $
3. $ k(x) = 7x $ → $ k' = 7 $

Apply step by step:

$$ h'(x) = 3[\cos(7x)]^2 \cdot (-\sin(7x)) \cdot 7 $$

$$ = -21 \cos^2(7x) \sin(7x) $$

---

```python
import sympy as sp
from sympy import symbols
from IPython.display import display, Math

x = sp.Symbol('x')

h1 = (x**2 + 3*x)**50
h2 = sp.exp(5*x**2)
h3 = sp.cos(7*x)**3

display(Math(fr"h_1'(x) = {sp.latex(sp.diff(h1, x))}"))
display(Math(fr"h_2'(x) = {sp.latex(sp.diff(h2, x))}"))
display(Math(fr"h_3'(x) = {sp.latex(sp.diff(h3, x))}"))
```

## Why This Matters in AI

Every neural network layer is a **composition**:

- $ z = w x + b $
- $ a = \sigma(z) $

Loss $ L(a) $ → to update $ w $, we need:

$$ \frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w} $$

**This is the Chain Rule** → **Backpropagation**.
