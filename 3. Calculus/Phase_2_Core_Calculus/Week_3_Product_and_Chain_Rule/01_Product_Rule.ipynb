## Introduction: Why Can't We Just Multiply Derivatives?

Let’s begin with a simple dream: you see $ h(x) = x^2 \cdot \sin(x) $. You think: *“I know how to differentiate $ x^2 $ and $ \sin(x) $. Why not just multiply their derivatives?”*

That’s the **trap**. Let’s **prove it fails** with a concrete example.

### The Test Case: $ h(x) = x^5 $

We know the **correct derivative** using the Power Rule:

$$ h'(x) = 5x^4 $$

Now, rewrite $ x^5 $ as a **product**:

$$ h(x) = x^2 \cdot x^3 $$

- Let $ f(x) = x^2 $ → $ f'(x) = 2x $
- Let $ g(x) = x^3 $ → $ g'(x) = 3x^2 $

Now apply the **trap method**:

$$ h'(x) \stackrel{?}{=} f'(x) \cdot g'(x) = 2x \cdot 3x^2 = 6x^3 $$

**Disaster.**

- **Correct**: $ 5x^4 $
- **Trap**: $ 6x^3 $

They are **not equal**. This proves:

> $$ \frac{d}{dx}[f(x)g(x)] \neq f'(x)g'(x) $$

We **cannot** just multiply derivatives. We need a **new rule**.

---

## Section 1: The Product Rule — The Correct Way

The **Product Rule** says:

> If $ h(x) = f(x) \cdot g(x) $, then:
> 
> $$ \boxed{h'(x) = f'(x)g(x) + f(x)g'(x)} $$

### The Mantra (Never Forget It)

> **“First times derivative of second… plus… second times derivative of first.”**

Let’s apply it to our test case:

- $ f(x) = x^2 $ → $ f'(x) = 2x $
- $ g(x) = x^3 $ → $ g'(x) = 3x^2 $

Apply rule:

$$ h'(x) = (2x)(x^3) + (x^2)(3x^2) $$

$$ = 2x^4 + 3x^4 = 5x^4 $$

**Perfect.** It matches the Power Rule.

---

### Why Does This Work? (Intuition)

Think of $ f(x)g(x) $ as **area of a rectangle** with sides $ f(x) $ and $ g(x) $.

When $ x $ changes by $ \Delta x $:

- One side changes by $ \Delta f $
- Other side changes by $ \Delta g $

The **change in area** is:

$$ \Delta A \approx f \cdot \Delta g + g \cdot \Delta f + \Delta f \cdot \Delta g $$

Divide by $ \Delta x $ and take limit → the extra term vanishes → **Product Rule**.

---

## Section 2: Real Examples — Step by Step

### Example 1: $ h(x) = x^2 \sin(x) $

- $ f(x) = x^2 $ → $ f'(x) = 2x $
- $ g(x) = \sin(x) $ → $ g'(x) = \cos(x) $

Apply:

$$ h'(x) = (2x)(\sin(x)) + (x^2)(\cos(x)) $$

$$ \boxed{h'(x) = 2x \sin(x) + x^2 \cos(x)} $$

---

### Example 2: $ h(x) = e^x \cos(x) $

- $ f(x) = e^x $ → $ f'(x) = e^x $
- $ g(x) = \cos(x) $ → $ g'(x) = -\sin(x) $

Apply:

$$ h'(x) = e^x \cos(x) + e^x (-\sin(x)) $$

$$ = e^x (\cos(x) - \sin(x)) $$

**Factoring is optional but elegant.**

---

### Example 3: Three Functions — $ h(x) = x^2 e^x \sin(x) $

Apply Product Rule **twice**:

Let $ u(x) = x^2 e^x $, $ v(x) = \sin(x) $

First, find $ u'(x) $:

- $ f = x^2 $ → $ f' = 2x $
- $ g = e^x $ → $ g' = e^x $
- $ u' = 2x e^x + x^2 e^x = e^x (2x + x^2) $

Now full derivative:

$$ h'(x) = u' v + u v' $$

$$ = [e^x (2x + x^2)] \sin(x) + [x^2 e^x] \cos(x) $$

**Leave as it is or factor $ e^x $**.

---

```python
import sympy as sp
from IPython.display import display, Math
sp.init_printing()

x = sp.symbols('x')
h1 = x**2 * sp.sin(x)
h2 = sp.exp(x) * sp.cos(x)
h3 = x**2 * sp.exp(x) * sp.sin(x)

display(Math(fr"h_1(x) = {sp.latex(h1)}"))
display(Math(fr"h_1'(x) = {sp.latex(sp.diff(h1, x))}"))

display(Math(fr"h_2(x) = {sp.latex(h2)}"))
display(Math(fr"h_2'(x) = {sp.latex(sp.diff(h2, x))}"))

display(Math(fr"h_3'(x) = {sp.latex(sp.diff(h3, x))}"))
```

## Quotient Rule (Division)

> $$ h(x) = \frac{f(x)}{g(x)} $$
> 
> $$ \boxed{h'(x) = \frac{f'g - fg'}{g^2}} $$

**Mantra**: *“Low d-high minus high d-low, square the bottom, away we go!”*

### Example: $ h(x) = \frac{\sin(x)}{x^2} $

- $ f = \sin(x) $ → $ f' = \cos(x) $
- $ g = x^2 $ → $ g' = 2x $

$$ h'(x) = \frac{\cos(x) \cdot x^2 - \sin(x) \cdot 2x}{(x^2)^2} $$

$$ = \frac{x^2 \cos(x) - 2x \sin(x)}{x^4} $$

---

## Why This Matters in AI

In neural networks, **activations** are often products:

- $ z = w \cdot x + b $
- $ a = \sigma(z) $

But later layers involve **products of activations and weights** → **Product Rule is essential in backprop**.

## Practice Problems (Full Walkthrough)

### Problem 1: $ h(x) = x^3 \ln(x) $

- $ f = x^3 $ → $ f' = 3x^2 $
- $ g = \ln(x) $ → $ g' = \frac{1}{x} $

$$ h'(x) = 3x^2 \ln(x) + x^3 \cdot \frac{1}{x} $$
$$ = 3x^2 \ln(x) + x^2 $$

$$ = x^2 (3 \ln(x) + 1) $$

### Problem 2: $ h(x) = \frac{x}{e^x} $

- $ f = x $ → $ f' = 1 $
- $ g = e^x $ → $ g' = e^x $

$$ h'(x) = \frac{1 \cdot e^x - x \cdot e^x}{(e^x)^2} = \frac{e^x (1 - x)}{e^{2x}} = \frac{1 - x}{e^x} $$

```python
p1 = x**3 * sp.ln(x)
p2 = x / sp.exp(x)

display(Math(fr"\text{{Problem 1: }} {sp.latex(sp.diff(p1, x))}"))
display(Math(fr"\text{{Problem 2: }} {sp.latex(sp.diff(p2, x))}"))
```
