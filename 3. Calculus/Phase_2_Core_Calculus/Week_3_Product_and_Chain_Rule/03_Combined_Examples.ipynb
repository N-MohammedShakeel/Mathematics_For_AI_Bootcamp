## Introduction: The Calculus Decision Tree — Your New Superpower

You now have **four powerful tools**:

1. **Sum Rule** → term by term
2. **Product Rule** → $ f'g + fg' $
3. **Quotient Rule** → $ \frac{f'g - fg'}{g^2} $
4. **Chain Rule** → $ f'(g(x)) \cdot g'(x) $

When you see a complex function, **don’t panic**. Ask:

> **What is the *main structure* of this function?**

This file is a **workshop** — we will solve **real-world, AI-relevant problems** by **identifying the top-level rule**, then **peeling layers** with **Chain Rule**, **Product Rule**, or **Quotient Rule**.

## Example 1: $ h(x) = e^{5x} \sin(x^2) $

### Step 1: Identify the Top-Level Structure

Look at the function:

$$ h(x) = e^{5x} \cdot \sin(x^2) $$

It is **two functions multiplied** → **Product Rule** is the **top-level rule**.

Define:

- $ f(x) = e^{5x} $
- $ g(x) = \sin(x^2) $

So:

$$ h'(x) = f'(x) g(x) + f(x) g'(x) $$

We need $ f'(x) $ and $ g'(x) $.

---

### Step 2: Find $ f'(x) $ — Chain Rule Inside

$ f(x) = e^{5x} $ is a **composite function**:

- **Outside**: $ e^u $ → derivative = $ e^u $
- **Inside**: $ u = 5x $ → derivative = $ 5 $

Chain Rule:

$$ f'(x) = e^{5x} \cdot 5 = 5e^{5x} $$

---

### Step 3: Find $ g'(x) $ — Chain Rule Again

$ g(x) = \sin(x^2) $ is also composite:

- **Outside**: $ \sin(v) $ → derivative = $ \cos(v) $
- **Inside**: $ v = x^2 $ → derivative = $ 2x $

Chain Rule:

$$ g'(x) = \cos(x^2) \cdot 2x = 2x \cos(x^2) $$

---

### Step 4: Assemble into Product Rule

Now plug in:

$$ h'(x) = (5e^{5x}) \cdot (\sin(x^2)) + (e^{5x}) \cdot (2x \cos(x^2)) $$

**Final Answer**:

$$ \boxed{h'(x) = 5e^{5x} \sin(x^2) + 2x e^{5x} \cos(x^2)} $$

You can factor $ e^{5x} $, but it’s optional.

---

```python
import sympy as sp
from IPython.display import display, Math
sp.init_printing()

x = sp.symbols('x')
h1 = sp.exp(5*x) * sp.sin(x**2)
dh1 = sp.diff(h1, x)

display(Math(fr"h(x) = {sp.latex(h1)}"))
display(Math(fr"h'(x) = {sp.latex(dh1)}"))
```

## Example 2: $ h(x) = \frac{(x^2 + 1)^3}{\ln(x)} $

### Step 1: Top-Level → Quotient Rule

$$ h(x) = \frac{\text{numerator}}{\text{denominator}} $$

- **High** $ f(x) = (x^2 + 1)^3 $
- **Low** $ g(x) = \ln(x) $

Quotient Rule:

$$ h'(x) = \frac{f'g - fg'}{g^2} $$

---

### Step 2: $ f'(x) $ — Chain Rule

$ f(x) = (x^2 + 1)^3 $

- Outside: $ u^3 $ → $ 3u^2 $
- Inside: $ u = x^2 + 1 $ → $ 2x $

$$ f'(x) = 3(x^2 + 1)^2 \cdot 2x = 6x (x^2 + 1)^2 $$

---

### Step 3: $ g'(x) $ — Base Rule

$ g(x) = \ln(x) $ → $ g'(x) = \frac{1}{x} $

---

### Step 4: Assemble

$$ h'(x) = \frac{ [6x (x^2 + 1)^2] \ln(x) - [(x^2 + 1)^3] \cdot \frac{1}{x} }{ [\ln(x)]^2 } $$

**Final Answer**:

$$ \boxed{ h'(x) = \frac{6x (x^2 + 1)^2 \ln(x) - \frac{(x^2 + 1)^3}{x}}{[\ln(x)]^2} } $$

---

```python
h2 = (x**2 + 1)**3 / sp.ln(x)
dh2 = sp.diff(h2, x)
display(Math(fr"h'(x) = {sp.latex(dh2)}"))
```

## Example 3: $ h(x) = \cos(x^3 e^x) $

### Step 1: Top-Level → Chain Rule

- **Outside**: $ \cos(u) $ → $ -\sin(u) $
- **Inside**: $ u = x^3 e^x $

$$ h'(x) = -\sin(x^3 e^x) \cdot \frac{d}{dx}(x^3 e^x) $$

---

### Step 2: Inside is Product → Product Rule

$ u(x) = x^3 \cdot e^x $

- $ f = x^3 $ → $ f' = 3x^2 $
- $ g = e^x $ → $ g' = e^x $

$$ u'(x) = 3x^2 e^x + x^3 e^x = e^x (3x^2 + x^3) $$

---

### Step 3: Final Answer

$$ h'(x) = -\sin(x^3 e^x) \cdot e^x (3x^2 + x^3) $$

**Final Answer**:

$$ \boxed{ h'(x) = -e^x (3x^2 + x^3) \sin(x^3 e^x) } $$

---

```python
h3 = sp.cos(x**3 * sp.exp(x))
dh3 = sp.diff(h3, x)
display(Math(fr"h'(x) = {sp.latex(dh3)}"))
```

## Practice Problems (Full Walkthrough)

### Problem 1: $ h(x) = x^2 \ln(\sin(x)) $

**Step 1**: Product → $ f = x^2 $, $ g = \ln(\sin(x)) $

**Step 2**: $ f' = 2x $

**Step 3**: $ g(x) = \ln(u) $, $ u = \sin(x) $ → $ g' = \frac{1}{u} \cdot \cos(x) = \frac{\cos(x)}{\sin(x)} $

**Answer**:

$$ h'(x) = 2x \ln(\sin(x)) + x^2 \cdot \frac{\cos(x)}{\sin(x)} $$

---

### Problem 2: $ h(x) = \frac{e^x}{\sqrt{x^2 + 1}} $

**Step 1**: Quotient

**Step 2**: $ f = e^x $ → $ f' = e^x $

**Step 3**: $ g = (x^2 + 1)^{1/2} $ → $ g' = \frac{1}{2}(x^2 + 1)^{-1/2} \cdot 2x = \frac{x}{\sqrt{x^2 + 1}} $

**Answer**:

$$ h'(x) = \frac{ e^x \sqrt{x^2 + 1} - e^x \cdot \frac{x}{\sqrt{x^2 + 1}} }{ x^2 + 1 } $$

---

```python
p1 = x**2 * sp.ln(sp.sin(x))
p2 = sp.exp(x) / sp.sqrt(x**2 + 1)

display(Math(fr"1. {sp.latex(sp.diff(p1, x))}"))
display(Math(fr"2. {sp.latex(sp.diff(p2, x))}"))
```
