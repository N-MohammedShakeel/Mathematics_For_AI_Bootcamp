## What is a Tangent Line?

Imagine you’re drawing a curve. At one point, you draw a **straight line** that **just touches** the curve — it doesn’t cross it.

That line is the **tangent line**.

It has the **same slope** as the curve at that exact point.

### We Need Two Things

1. **A point** $ (x_1, y_1) $
2. **The slope** $ m = f'(x_1) $

Then use:

$$ y - y_1 = m(x - x_1) $$

### Example: $ f(x) = x^3 - 2x + 1 $ at $ x = 2 $

**Step 1: Find the point**

- $ x_1 = 2 $
- $ y_1 = f(2) = 8 - 4 + 1 = 5 $
- Point: $ (2, 5) $

**Step 2: Find the slope**

- $ f'(x) = 3x^2 - 2 $
- $ m = f'(2) = 12 - 2 = 10 $

**Step 3: Write the equation**

$$ y - 5 = 10(x - 2) $$

$$ y = 10x - 15 $$

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(0, 3, 400)
f = x_vals**3 - 2*x_vals + 1
tangent = 10*x_vals - 15

plt.plot(x_vals, f, 'b-', label='$f(x)$')
plt.plot(x_vals, tangent, 'r--', label='Tangent')
plt.scatter([2], [5], c='red', zorder=5)
plt.title('Tangent Line at x=2')
plt.legend(); plt.grid(); plt.show()
```

## Normal Line: Perpendicular to Tangent

> **Slope of normal = negative reciprocal of tangent slope**

$$ m_n = -\frac{1}{m_t} $$

- Tangent slope = 10 → Normal slope = $ -\frac{1}{10} $
- Equation: 

$$ y - 5 = -\frac{1}{10}(x - 2) $$

```python
normal = -0.1*(x_vals - 2) + 5

plt.plot(x_vals, f, 'b-', label='$f(x)$')
plt.plot(x_vals, normal, 'g--', label='Normal Line')
plt.scatter([2], [5], c='red', zorder=5)
plt.title('Normal Line at x=2')
plt.legend(); plt.grid(); plt.show()
```

## General Process for Tangent & Normal

### **Tangent Line Steps**
1. Find the point: compute $ f(a) $
2. Find the derivative $ f'(x) $
3. Compute slope at point: $ m = f'(a) $
4. Apply formula: 

$$ y - f(a) = m(x - a) $$

---

### **Normal Line Steps**
1. Find tangent slope $ m_t $
2. Compute normal slope: 

$$ m_n = -\frac{1}{m_t} $$

3. Apply formula:

$$ y - f(a) = m_n(x - a) $$

## Why Tangent & Normal Matter

### **Applications**
- **Physics**: motion, forces → normal to trajectory
- **Computer Graphics**: surface reflection uses **normal vectors**
- **Engineering**: stress, curvature, contact forces
- **Robotics**: orientation & control
- **Machine Learning**: optimization follows tangent/gradient direction

Understanding tangent/normal lines is foundational for geometry, physics simulations, and ML optimization theory.
