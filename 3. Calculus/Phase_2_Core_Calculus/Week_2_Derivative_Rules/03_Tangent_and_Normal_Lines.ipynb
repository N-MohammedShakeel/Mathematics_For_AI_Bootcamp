## What is a Tangent Line?

Imagine you’re drawing a curve. At one point, you draw a **straight line** that **just touches** the curve — it doesn’t cross it.

That line is the **tangent line**.

It has the **same slope** as the curve at that exact point.

### We Need Two Things

1. **A point** $ (x_1, y_1) $
2. **The slope** $ m = f'(x_1) $

Then use:

$$ y - y_1 = m(x - x_1) $$

### Example: $ f(x) = x^3 - 2x + 1 $ at $ x = 2 $

**Step 1: Find the point**

- $ x_1 = 2 $
- $ y_1 = f(2) = 8 - 4 + 1 = 5 $
- Point: $ (2, 5) $

**Step 2: Find the slope**

- $ f'(x) = 3x^2 - 2 $
- $ m = f'(2) = 12 - 2 = 10 $

**Step 3: Write the equation**

$$ y - 5 = 10(x - 2) $$

$$ y = 10x - 15 $$

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(0, 3, 400)
f = x_vals**3 - 2*x_vals + 1
tangent = 10*x_vals - 15

plt.plot(x_vals, f, 'b-', label='$f(x)$')
plt.plot(x_vals, tangent, 'r--', label='Tangent')
plt.scatter([2], [5], c='red', zorder=5)
plt.title('Tangent Line at x=2')
plt.legend(); plt.grid(); plt.show()
```

## Normal Line: Perpendicular to Tangent

> **Slope of normal = negative reciprocal**

$$ m_n = -\frac{1}{m_t} $$

- Tangent slope = 10 → Normal slope = $ -\frac{1}{10} $
- Equation: $ y - 5 = -\frac{1}{10}(x - 2) $

## Why It Matters

- **Physics**: Gravity pulls along **normal** to orbit
- **Graphics**: Light reflects off **normal vector**
