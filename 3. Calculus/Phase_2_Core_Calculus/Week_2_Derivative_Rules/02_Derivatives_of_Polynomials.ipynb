## What is a Polynomial?

Let’s start with a simple idea: a **polynomial** is a function made by **adding or subtracting** terms that look like:

$$ \text{constant} \times x^{\text{whole number}} $$

For example:

$$ P(x) = 3x^4 - 2x^3 + 5x^2 + x - 7 $$

Each term is **$ a \cdot x^n $**, where $ n $ is a **non-negative integer** (0, 1, 2, 3, ...).

### Why Do We Care?

Polynomials are **smooth**, **predictable**, and **easy to differentiate**.

In AI, **loss functions** are often polynomials in the weights. So mastering polynomials = mastering optimization.

---

## The Term-by-Term Method: Your New Superpower

Because of the **Sum Rule**, we can differentiate a polynomial **one term at a time**.

Let’s do it slowly.

### Example 1: $ f(x) = 3x^2 + 5x - 7 $

We break it into three terms:

1. $ 3x^2 $
   - Constant Multiple: $ c = 3 $
   - Power Rule: $ n = 2 $ → $ 2x $
   - $ 3 \cdot 2x = 6x $

2. $ 5x $
   - $ 5x^1 $ → $ 5 \cdot 1 \cdot x^0 = 5 $

3. $ -7 $
   - Constant → $ 0 $

**Final Answer**:

$$ f'(x) = 6x + 5 $$

---

### Example 2: Higher-Order Polynomial

$$ P(x) = -x^5 + 10x^4 + \frac{1}{3}x^3 - \frac{3}{4}x^2 + 15x - 1 $$

Let’s go term by term:

1. $ -x^5 $ → $ -5x^4 $
2. $ 10x^4 $ → $ 40x^3 $
3. $ \frac{1}{3}x^3 $ → $ \frac{1}{3} \cdot 3x^2 = x^2 $
4. $ -\frac{3}{4}x^2 $ → $ -\frac{3}{4} \cdot 2x = -\frac{3}{2}x $
5. $ 15x $ → $ 15 $
6. $ -1 $ → $ 0 $

**Final Answer**:

$$ P'(x) = -5x^4 + 40x^3 + x^2 - \frac{3}{2}x + 15 $$

```python
import sympy as sp
from sympy import symbols
from IPython.display import display, Math

x = sp.Symbol('x')

P = -x**5 + 10*x**4 + sp.Rational(1,3)*x**3 - sp.Rational(3,4)*x**2 + 15*x - 1
dP = sp.diff(P, x)
display(Math(fr"P'(x) = {sp.latex(dP)}"))
```

## Higher-Order Derivatives: The Derivative of the Derivative

The derivative $ f'(x) $ is a **new function**.

So we can take its derivative → **second derivative** $ f''(x) $

- $ f'(x) $: slope (velocity)
- $ f''(x) $: concavity (acceleration)

### Example

$$ f(x) = 4x^3 - 5x^2 + 2x - 1 $$

- $ f'(x) = 12x^2 - 10x + 2 $
- $ f''(x) = 24x - 10 $
- $ f'''(x) = 24 $
- $ f^{(4)}(x) = 0 $

**After degree $ n $, all higher derivatives are zero.**

```python
f = 4*x**3 - 5*x**2 + 2*x - 1
print("1st:", sp.diff(f, x))
print("2nd:", sp.diff(f, x, 2))
print("3rd:", sp.diff(f, x, 3))
print("4th:", sp.diff(f, x, 4))
```

## AI: Finding the Minimum Loss

In AI, we want to **minimize** the loss:

$$ L(w) = (w - 2)^2 + 1 $$

Expand:

$$ L(w) = w^2 - 4w + 5 $$

1. $ L'(w) = 2w - 4 $
2. Set $ L'(w) = 0 $ → $ w = 2 $
3. $ L''(w) = 2 > 0 $ → **minimum**

**This is gradient descent in math.**

## Practice

1. $ P(x) = x^4 - 2x^3 + x $ → find $ P''(x) $
2. Loss: $ L(w) = (w-3)^2 + 2 $ → find minimum

**Solutions**

```python
# 1
P = x**4 - 2*x**3 + x
display(sp.diff(P, x, 2))

# 2
L = (x-3)**2 + 2
dL = sp.diff(L, x)
crit = sp.solve(dL, x)[0]
d2L = sp.diff(L, x, 2)
print(f"Min at w = {crit}, L'' = {d2L} > 0 → minimum")
```
