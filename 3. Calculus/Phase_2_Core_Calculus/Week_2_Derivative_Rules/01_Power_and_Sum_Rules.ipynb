## Introduction: Why Do We Need Shortcuts?

Imagine you’re in a car, and you want to know **how fast you’re going** at every single moment.

You *could* calculate your average speed over the last 0.0001 seconds — that’s the **limit definition** we learned last week.

But what if you had a **speedometer** that just *told* you the speed instantly?

That’s what **derivative rules** are — **speedometers for functions**.

Last week, we proved that the derivative of $ f(x) = x^2 $ is $ f'(x) = 2x $ using this long limit:

$$ f'(x) = \lim_{h \to 0} \frac{(x+h)^2 - x^2}{h} = \lim_{h \to 0} \frac{2xh + h^2}{h} = 2x $$

It worked. But now imagine doing that for $ x^8 $.

You’d have to expand $ (x+h)^8 $ — that’s **9 terms** of algebra. Then subtract $ x^8 $. Then divide by $ h $. Then cancel. Then take the limit.

**That’s insane.**

So mathematicians said: *“Let’s do this hard work **once**, prove a pattern, and then use it forever.”*

That pattern is the **Power Rule**.

---

## Section 1: The Power Rule — The Most Important Rule in Calculus

Let’s start with a simple question:

> What happens to the derivative every time we see $ x $ raised to a power?

We already know:

- $ f(x) = x^2 $ → $ f'(x) = 2x $
- $ f(x) = x^3 $ → $ f'(x) = 3x^2 $

Do you see the pattern?

- The **exponent comes down** and becomes a **multiplier**.
- Then the exponent **decreases by 1**.

So for $ x^n $, the derivative should be $ n x^{n-1} $.

### The Power Rule (Official)

> **For any function of the form $ f(x) = x^n $ (where $ n $ is any real number), the derivative is:**
>
> $$ \boxed{f'(x) = n \cdot x^{n-1}} $$

Let’s test it with $ x^8 $:

- $ n = 8 $
- Bring down 8 → multiply
- Subtract 1 from power → $ x^7 $
- $ f'(x) = 8x^7 $

**One second. No limits. No algebra hell.**

---

### Special Case 1: What About Constants?

What is the derivative of $ f(x) = 5 $? (A flat line)

The slope of a flat line is **zero**. So $ f'(x) = 0 $.

But can we use the Power Rule?

Yes! Rewrite $ 5 $ as $ 5x^0 $ (because $ x^0 = 1 $).

- $ n = 0 $
- $ f'(x) = 0 \cdot x^{0-1} = 0 \cdot x^{-1} = 0 $

**It works!**

> **Constant Rule**: $ \frac{d}{dx}[c] = 0 $

---

### Special Case 2: What About $ f(x) = x $?

This is the line $ y = x $. Slope is 1 everywhere.

Rewrite: $ x = x^1 $

- $ n = 1 $
- $ f'(x) = 1 \cdot x^{1-1} = 1 \cdot x^0 = 1 \cdot 1 = 1 $

**Perfect.**

> **Rule**: $ \frac{d}{dx}[x] = 1 $

---

### What About Roots and Fractions?

The Power Rule works for **any real number** $ n $ — including fractions and negatives.

#### Example 1: $ f(x) = \sqrt{x} $

Rewrite: $ \sqrt{x} = x^{1/2} $

- $ n = \frac{1}{2} $
- $ f'(x) = \frac{1}{2} x^{1/2 - 1} = \frac{1}{2} x^{-1/2} $
- $ x^{-1/2} = \frac{1}{x^{1/2}} = \frac{1}{\sqrt{x}} $
- $ f'(x) = \frac{1}{2\sqrt{x}} $

**Common and important.**

#### Example 2: $ f(x) = \frac{1}{x} $

Rewrite: $ \frac{1}{x} = x^{-1} $

- $ n = -1 $
- $ f'(x) = -1 \cdot x^{-1-1} = -x^{-2} = -\frac{1}{x^2} $

**Another classic.**

---

## Section 2: The Constant Multiple Rule

Now, what if we have $ 5x^2 $? Or $ \frac{7}{x^3} $?

We don’t want to redo the limit proof.

Instead, we say:

> **If a function is multiplied by a constant, just multiply the derivative by that same constant.**

### The Rule

$$ \frac{d}{dx}[c \cdot f(x)] = c \cdot f'(x) $$

#### Example: $ g(x) = 5x^2 $

- $ c = 5 $, $ f(x) = x^2 $
- $ f'(x) = 2x $
- $ g'(x) = 5 \cdot 2x = 10x $

#### Example: $ g(x) = \frac{7}{x^3} = 7x^{-3} $

- $ c = 7 $, $ f(x) = x^{-3} $
- $ f'(x) = -3x^{-4} $
- $ g'(x) = 7 \cdot (-3x^{-4}) = -21x^{-4} = -\frac{21}{x^4} $

---

## Section 3: The Sum and Difference Rule

Now, what if we have multiple terms?

> **The derivative of a sum is the sum of the derivatives.**

### The Rule

$$ \frac{d}{dx}[f(x) + g(x)] = f'(x) + g'(x) $$

$$ \frac{d}{dx}[f(x) - g(x)] = f'(x) - g'(x) $$

This means: **Differentiate term by term.**

---

## Full Example: Putting It All Together

Let’s differentiate:

$$ h(x) = 6x^4 - 2x^3 + x - 10 $$

We go **term by term**:

1. **$ 6x^4 $**
   - Constant Multiple: $ c = 6 $
   - Power Rule: $ n = 4 $ → $ 4x^3 $
   - $ 6 \cdot 4x^3 = 24x^3 $

2. **$ -2x^3 $**
   - $ c = -2 $, $ n = 3 $ → $ 3x^2 $
   - $ -2 \cdot 3x^2 = -6x^2 $

3. **$ x $**
   - $ x = x^1 $ → $ 1 \cdot x^0 = 1 $

4. **$ -10 $**
   - Constant → $ 0 $

**Final Answer**:

$$ h'(x) = 24x^3 - 6x^2 + 1 $$

---

```python
import sympy as sp
from IPython.display import display, Math
sp.init_printing()

x = sp.symbols('x')
h = 6*x**4 - 2*x**3 + x - 10
dh = sp.diff(h, x)

display(Math(fr"h(x) = {sp.latex(h)}"))
display(Math(fr"h'(x) = {sp.latex(dh)}"))
```

## Why This Matters in AI

In AI, the **loss function** is often a polynomial in the weights $ w $:

$$ L(w) = (w - 3)^2 + 5 $$

To find the **best weight**, we set $ L'(w) = 0 $.

Using Power + Sum Rule:

- $ L(w) = w^2 - 6w + 14 $
- $ L'(w) = 2w - 6 $
- $ 2w - 6 = 0 $ → $ w = 3 $

**This is how AI learns.**

## Practice Problems (Full Walkthrough)

### Problem 1

$$ f(x) = 4x^5 - 3\sqrt{x} + \frac{2}{x} $$

**Step 1**: Rewrite

- $ \sqrt{x} = x^{1/2} $
- $ \frac{2}{x} = 2x^{-1} $

**Step 2**: Term by term

- $ 4x^5 $ → $ 20x^4 $
- $ -3x^{1/2} $ → $ -\frac{3}{2} x^{-1/2} $
- $ 2x^{-1} $ → $ -2x^{-2} $

**Answer**:

$$ f'(x) = 20x^4 - \frac{3}{2\sqrt{x}} - \frac{2}{x^2} $$

```python
f = 4*x**5 - 3*sp.sqrt(x) + 2/x
df = sp.diff(f, x)
display(Math(fr"f'(x) = {sp.latex(df)}"))
```
