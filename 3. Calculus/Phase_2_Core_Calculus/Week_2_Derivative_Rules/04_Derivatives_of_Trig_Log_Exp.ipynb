## Beyond Polynomials

Power Rule fails on $ \sin(x) $, $ e^x $, $ \ln(x) $

→ New **lookup table rules**

## 1. $ e^x $: The Magic Function

The number $ e \approx 2.718 $ is special.

The function $ e^x $ has a **unique property**:

> **Its slope at any point equals its value at that point.**

So:

$$ \frac{d}{dx}[e^x] = e^x $$

### General: $ a^x $

$$ \frac{d}{dx}[a^x] = a^x \ln(a) $$

## 2. Natural Log: $ \ln(x) $

$ \ln(x) $ is the inverse of $ e^x $.

> **$ \frac{d}{dx}[\ln(x)] = \frac{1}{x} $**

### General Log

$$ \frac{d}{dx}[\log_a(x)] = \frac{1}{x \ln(a)} $$

## 3. Trig Functions (Radians!)

| Function | Derivative |
|--------|----------|
| $ \sin(x) $ | $ \cos(x) $ |
| $ \cos(x) $ | $ -\sin(x) $ |
| $ \tan(x) $ | $ \sec^2(x) $ |

> **Must be in radians**

```python
import sympy as sp

x = sp.Symbol('x')

funcs = [sp.exp(x), sp.ln(x), sp.sin(x), sp.cos(x)]
for f in funcs:
    df = sp.diff(f, x)
    display(sp.Eq(sp.Derivative(f), df))
```

## AI Applications

- **Sigmoid**: $ \sigma(x) = \frac{1}{1 + e^{-x}} $ → needs $ e^x $
- **Cross-Entropy Loss**: needs $ \ln(x) $
- **Softmax**: uses $ e^x $

> **Backpropagation = these rules in code**
