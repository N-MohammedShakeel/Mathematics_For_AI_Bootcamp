## What We’re Doing Today

We know the **derivative** is the **instantaneous slope** of a function.

But how do we **calculate** it?

Today, we’ll learn the **one true formula** that lets us find the derivative of **any function** — even if we don’t know any shortcuts.

We’ll use it on:

- $ f(x) = x^2 $
- $ f(x) = 3x + 5 $

And we’ll do **every single step** — no magic, no skipping.

---

## 1. The Most Important Formula in Calculus

Remember from **Phase 1**:

> We found the **average slope** between two points using:
> $$ \frac{f(x+h) - f(x)}{h} $$

And we said: if we make $ h $ **very small**, this becomes the **instant slope**.

Now, we make that idea **exact** using a **limit**:

$$ \boxed{f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}} $$

### What This Means in English

> “Take two points on the curve, very close together. Find the slope between them. Then imagine making the distance between them **zero**, but in a smart way that still gives a real answer. That answer is the derivative.”

This is called **finding the derivative from first principles**.

It’s like **proving** the slope, not just guessing it.

---

## 2. Example 1: $ f(x) = x^2 $ → Prove $ f'(x) = 2x $

Let’s go **slowly**, step by step.

### Step 1: Write the Definition

$$ f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h} $$

### Step 2: Replace $ f(x) $

$$ f(x) = x^2 $$

### Step 3: Compute $ f(x+h) $

$$ f(x+h) = (x + h)^2 $$

Expand carefully:

$$ (x + h)^2 = x^2 + 2xh + h^2 $$

### Step 4: Plug into the Formula

$$ f'(x) = \lim_{h \to 0} \frac{(x^2 + 2xh + h^2) - x^2}{h} $$

### Step 5: Simplify the Top

$$ = \lim_{h \to 0} \frac{2xh + h^2}{h} $$

### Step 6: Factor $ h $ from Numerator

$$ = \lim_{h \to 0} \frac{h(2x + h)}{h} $$

### Step 7: Cancel $ h $ (Safe because $ h \neq 0 $ in limit)

$$ = \lim_{h \to 0} (2x + h) $$

### Step 8: Take the Limit

$$ f'(x) = 2x + 0 = 2x $$

**We proved it!**

---

## Let’s See It in Code

```python
import sympy as sp
from IPython.display import display, Math
sp.init_printing()

x, h = sp.symbols('x h')
f = x**2

# Step by step
fxh = (x + h)**2
numerator = fxh - f
quotient = numerator / h
simplified = sp.simplify(quotient)
derivative = sp.limit(simplified, h, 0)

display(Math(fr"f(x+h) = {sp.latex(fxh)}"))
display(Math(fr"\frac{{f(x+h)-f(x)}}{{h}} = {sp.latex(quotient)}"))
display(Math(fr"= {sp.latex(simplified)}"))
display(Math(fr"\therefore f'(x) = {derivative}"))
```

## 3. Example 2: $ f(x) = 3x + 5 $ → Constant Slope

We know the slope is 3. Let’s **prove** it.

### Step 1: $ f(x+h) = 3(x+h) + 5 = 3x + 3h + 5 $

### Step 2:

$$ \frac{(3x + 3h + 5) - (3x + 5)}{h} = \frac{3h}{h} = 3 $$

### Step 3:

$$ \lim_{h \to 0} 3 = 3 $$

$$ \boxed{f'(x) = 3} $$

**Perfect match!**

## 4. Why This Matters in AI

Every **fast rule** (Power Rule, Chain Rule) comes from this limit.

In **deep learning**, we compute millions of gradients — all based on this idea.

> **This limit is the DNA of backpropagation.**

Use the limit to find $ f'(x) $ for:

2. $ f(x) = 4x - 7 $

**Solutions below**

```python
# Solution 1: x^3
f = x**3
diff = ((x+h)**3 - x**3) / h
expanded = sp.expand(diff)
derivative = sp.limit(expanded, h, 0)
display(Math(fr"f'(x) = x^3 \rightarrow {derivative}"))

# Solution 2: 4x - 7
f = 4*x - 7
derivative = sp.limit((f.subs(x,x+h) - f)/h, h, 0)
display(Math(fr"f'(x) = 4x - 7 \rightarrow {derivative}"))
```
