## 1. The Problem: Slope on a Curve

Let’s start with something you already know from **Phase 1**:

> **For a straight line**, the **slope is constant**.

Example: $ f(x) = 3x + 5 $

- From $ x=0 $ to $ x=1 $ → $ y $ goes from 5 to 8 → slope = 3
- From $ x=10 $ to $ x=11 $ → $ y $ goes from 35 to 38 → slope = 3

**Always 3**. Easy.

---

Now look at a **curve**: $ f(x) = x^2 $

| At $ x = -2 $ | $ f(x) = 4 $ | The curve is **steep and going down** |
|---------------|-------------|-------------------------------------|
| At $ x = 0 $  | $ f(x) = 0 $ | The curve is **completely flat**   |
| At $ x = 2 $  | $ f(x) = 4 $ | The curve is **steep and going up** |

**The slope keeps changing!**

So how do we answer:

> **"What is the slope of $ f(x) = x^2 $?"**

We can’t say “it’s 4” or “it’s 0” — because it depends on **where** you are.

**This is the big idea of calculus**:

> **We need a new function that tells us the slope at every point.**

That function is called the **derivative**.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8')

# Plot f(x) = x²
x = np.linspace(-3, 3, 400)
y = x**2

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=3, label='$f(x) = x^2$')

# Mark key points
plt.scatter([-2, 0, 2], [4, 0, 4], color='red', s=100, zorder=5)
plt.text(-2.1, 4.5, 'Steep Down', fontsize=14, ha='center', bbox=dict(boxstyle='round', facecolor='yellow'))
plt.text(0, 0.8, 'Flat!', fontsize=14, ha='center', bbox=dict(boxstyle='round', facecolor='lightgreen'))
plt.text(2.1, 4.5, 'Steep Up', fontsize=14, ha='center', bbox=dict(boxstyle='round', facecolor='yellow'))

plt.title('The Slope of $f(x) = x^2$ Changes!', fontsize=16)
plt.xlabel('x', fontsize=14)
plt.ylabel('f(x)', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True)
plt.show()
```

## 2. The Speedometer Analogy (Best Way to Understand Derivative)

Imagine you’re in a car:

- $ f(t) $ = your **position** (distance from home) at time $ t $
- You drive 1 hour → travel 60 km → **average speed** = 60 km/h

But at **any moment**, you look at your **speedometer**:

- 30 km/h
- 90 km/h
- 0 km/h (at red light)

**That number on the speedometer is your instantaneous speed.**

| Function | Meaning |
|--------|-------|
| $ f(t) $ | **Position** at time $ t $ |
| $ f'(t) $ | **Speed** at time $ t $ |

**The derivative is the speedometer of a function.**

- $ f'(t) > 0 $ → moving **forward**
- $ f'(t) < 0 $ → moving **backward**
- $ f'(t) = 0 $ → **stopped**

---

**For $ f(x) = x^2 $**, we will soon learn:

$$ \boxed{f'(x) = 2x} $$

Let’s check:

| $ x $ | $ f'(x) = 2x $ | Meaning |
|------|----------------|--------|
| 0    | 0              | Flat (stopped) |
| 2    | 4              | Steep up (fast increase) |
| -2   | -4             | Steep down (fast decrease) |

**It matches perfectly!**

## 3. Notation: How to Write “Derivative”

There are **two standard ways** — both mean the **same thing**.

### 1. Lagrange Notation: $ f'(x) $

- Read: **"f prime of x"**
- Clean and simple
- Most common in AI/ML

### 2. Leibniz Notation: $ \frac{dy}{dx} $

- Read: **"dee y dee x"**
- Looks like a fraction → reminds us of **slope**

> It comes from:
> $$ \frac{\Delta y}{\Delta x} \text{ (average slope)} \rightarrow \frac{dy}{dx} \text{ (instant slope)} $$

Both are correct:

$$ f'(x) = \frac{dy}{dx} $$

## 4. Why AI Needs Derivatives

In AI, we train models by **minimizing error** (loss).

- Loss $ L(w) $ = how wrong the model is
- We want the **lowest point** (minimum)

**How do we find the bottom of a curve?**

→ **Where the slope is zero!**

So we:

1. Pick random weights $ w $
2. Compute **slope** $ L'(w) $
3. If slope > 0 → go **left**
4. If slope < 0 → go **right**
5. Repeat until $ L'(w) \approx 0 $

This is **Gradient Descent** — the **engine of AI**.

> **No derivative = No learning**
