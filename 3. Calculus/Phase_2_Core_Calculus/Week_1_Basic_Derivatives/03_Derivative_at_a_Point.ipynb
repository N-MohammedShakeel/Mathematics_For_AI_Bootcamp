## Two Types of Questions

We now know how to find the **general derivative** $ f'(x) $.

But sometimes, we only care about the **slope at one point**.

| Question | Answer Type | Example |
|--------|-------------|--------|
| What is the slope formula? | **Function** | $ f'(x) = 2x $ |
| What is the slope at $ x = 3 $? | **Number** | $ f'(3) = 6 $ |

Today: **How to find $ f'(3) $** — two ways.

## 1. Method 1: Use the General Formula (Best for AI)

If you already know $ f'(x) = 2x $, just **plug in** $ x = 3 $:

$$ f'(3) = 2(3) = 6 $$

**Fast. Simple. Done.**

This is what AI does — compute $ f'(x) $ once, then evaluate at any point.

## 2. Method 2: Direct Limit at Point $ c $

Alternative formula:

$$ \boxed{f'(c) = \lim_{x \to c} \frac{f(x) - f(c)}{x - c}} $$

### What This Means

> “Take a point $ c $, and another point $ x $ very close to it. Find the slope between them. Then let $ x $ get closer and closer to $ c $. The final slope is $ f'(c) $.”

It’s just the **secant slope** with a limit.

## 3. Example: $ f(x) = x^2 $ at $ x = 3 $

### Method 1 (Fast)

$$ f'(x) = 2x \quad \Rightarrow \quad f'(3) = 6 $$

### Method 2 (From Scratch)

$$ f'(3) = \lim_{x \to 3} \frac{x^2 - 3^2}{x - 3} $$

$$ = \lim_{x \to 3} \frac{x^2 - 9}{x - 3} $$

Factor numerator:

$$ = \lim_{x \to 3} \frac{(x-3)(x+3)}{x - 3} $$

Cancel $ x-3 $:

$$ = \lim_{x \to 3} (x + 3) $$

Plug in $ x = 3 $:

$$ f'(3) = 3 + 3 = 6 $$

**Same answer!**

```python
x = sp.symbols('x')
f = x**2
c = 3
limit_val = sp.limit((f - f.subs(x,c)) / (x - c), x, c)

display(Math(fr"f'({c}) = \lim_{{x \to {c}}} \frac{{f(x) - f({c})}}{{x - {c}}} = {limit_val}"))
```

## 4. When You CAN’T Take a Derivative

A function is **differentiable at $ c $** if $ f'(c) $ exists.

### Not Differentiable When:

1. **Sharp Corner** → two different slopes meet
   - Example: $ f(x) = |x| $ at $ x = 0 $

2. **Jump or Hole** → function breaks

3. **Vertical Tangent** → slope goes to infinity

**Golden Rule**: If differentiable → must be **continuous**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 400)
y = np.abs(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'purple', linewidth=3, label='$f(x) = |x|$')
plt.scatter([0], [0], c='red', s=120, zorder=5)
plt.text(0.3, 0.4, 'Sharp Corner!\nNo Derivative Here', fontsize=14, 
         bbox=dict(boxstyle='round,pad=0.5', facecolor='pink'))
plt.title('Not Differentiable at x = 0')
plt.xlabel('x'); plt.ylabel('f(x)')
plt.legend(); plt.grid(True)
plt.show()
```

## 5. Why AI Needs Smooth Functions

- Gradient descent **follows the slope**
- If slope doesn’t exist → **algorithm breaks**
- ReLU has a corner → we use **subgradient**
- Loss functions must be **smooth and differentiable**

> **No derivative = No learning**

## Practice: Find $ f'(2) $

For $ f(x) = x^3 - x $

**Answer below**

```python
f = x**3 - x
df = sp.diff(f, x)  # We'll learn this soon!
value = df.subs(x, 2)
display(Math(fr"f(x) = x^3 - x \quad \Rightarrow \quad f'(x) = {df} \quad \Rightarrow \quad f'(2) = {value}"))
```
