# 18. Positive Definite Matrices & Quadratic Forms

Welcome to **Positive Definite Matrices and Quadratic Forms**!

In Machine Learning and Optimization, understanding matrix definiteness is critical:
- It determines whether a multi-dimensional loss function has a **unique global minimum** (convex bowl) vs a saddle point.
- It defines the structure of **Covariance Matrices** in statistics and Gaussian models.
- It enables fast, numerically stable matrix factorizations like the **Cholesky Decomposition**.

---

## 1. Quadratic Forms

### Definition:
Given a symmetric matrix $\mathbf{A} \in \mathbb{R}^{n 	imes n}$ and a vector $\mathbf{x} \in \mathbb{R}^n$, the function:
$$q(\mathbf{x}) = \mathbf{x}^T \mathbf{A} \mathbf{x} = \sum_{i=1}^n \sum_{j=1}^n A_{ij} x_i x_j$$
is called a **Quadratic Form**.

### 2D Example:
For $\mathbf{A} = egin{bmatrix} a & b \ b & c \end{bmatrix}$ and $\mathbf{x} = egin{bmatrix} x_1 \ x_2 \end{bmatrix}$:
$$q(\mathbf{x}) = egin{bmatrix} x_1 & x_2 \end{bmatrix} egin{bmatrix} a & b \ b & c \end{bmatrix} egin{bmatrix} x_1 \ x_2 \end{bmatrix} = a x_1^2 + 2b x_1 x_2 + c x_2^2$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Compute quadratic form for a 2x2 matrix
A = np.array([[3.0, 1.0], 
              [1.0, 2.0]])
x = np.array([2.0, -1.0])

# x^T A x
q_val = x.T @ A @ x
print(f"Vector x: {x}")
print(f"Matrix A:\n{A}")
print(f"Quadratic Form x^T A x = {q_val}")
```

---

## 2. Matrix Definiteness Classification

Depending on the sign of $\mathbf{x}^T \mathbf{A} \mathbf{x}$ for all non-zero vectors $\mathbf{x} 
eq \mathbf{0}$:

| Classification | Condition for all $\mathbf{x} 
eq \mathbf{0}$ | Eigenvalues $\lambda_i$ | Geometric Shape |
| :--- | :--- | :--- | :--- |
| **Positive Definite (PD)** | $\mathbf{x}^T \mathbf{A} \mathbf{x} > 0$ | All $\lambda_i > 0$ | Upward bowl (strict minimum at $\mathbf{0}$) |
| **Positive Semi-Definite (PSD)** | $\mathbf{x}^T \mathbf{A} \mathbf{x} \geq 0$ | All $\lambda_i \geq 0$ | Trough / Valley (flat bottom) |
| **Negative Definite (ND)** | $\mathbf{x}^T \mathbf{A} \mathbf{x} < 0$ | All $\lambda_i < 0$ | Downward bowl (strict maximum at $\mathbf{0}$) |
| **Indefinite** | Takes both positive and negative values | Mixed signs ($\lambda_1 > 0, \lambda_2 < 0$) | **Saddle surface** (min in one direction, max in another) |

## 3. Geometric Visualization of Quadratic Forms


```python
# 3D Visualizer of Quadratic Forms
x1 = np.linspace(-3, 3, 100)
x2 = np.linspace(-3, 3, 100)
X1, X2 = np.meshgrid(x1, x2)

# 1. Positive Definite Matrix: [[2, 0], [0, 2]] -> f = 2*x1^2 + 2*x2^2
Z_pd = 2 * X1**2 + 2 * X2**2

# 2. Indefinite Matrix: [[2, 0], [0, -2]] -> f = 2*x1^2 - 2*x2^2 (Saddle point)
Z_indef = 2 * X1**2 - 2 * X2**2

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.plot_surface(X1, X2, Z_pd, cmap='viridis', alpha=0.8)
ax1.set_title("Positive Definite: Strict Global Minimum (Bowl)")
ax1.set_xlabel("x1"); ax1.set_ylabel("x2"); ax1.set_zlabel("q(x)")

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.plot_surface(X1, X2, Z_indef, cmap='coolwarm', alpha=0.8)
ax2.set_title("Indefinite Matrix: Saddle Point")
ax2.set_xlabel("x1"); ax2.set_ylabel("x2"); ax2.set_zlabel("q(x)")

plt.tight_layout()
plt.show()
```

## 4. Covariance Matrices in AI/ML

### Why are Covariance Matrices always Positive Semi-Definite (PSD)?
Let $\mathbf{X} \in \mathbb{R}^{N 	imes D}$ be a centered data matrix (mean = 0).
The sample covariance matrix is:
$$\mathbf{\Sigma} = rac{1}{N} \mathbf{X}^T \mathbf{X}$$

For any arbitrary vector $\mathbf{v} \in \mathbb{R}^D$:
$$\mathbf{v}^T \mathbf{\Sigma} \mathbf{v} = \mathbf{v}^T \left(rac{1}{N} \mathbf{X}^T \mathbf{X}ight) \mathbf{v} = rac{1}{N} (\mathbf{X}\mathbf{v})^T (\mathbf{X}\mathbf{v}) = rac{1}{N} \|\mathbf{X}\mathbf{v}\|_2^2 \geq 0$$

Since the squared Euclidean norm $\|\mathbf{X}\mathbf{v}\|_2^2$ is always $\geq 0$, **the covariance matrix is guaranteed to be Positive Semi-Definite!**


```python
# Generate random 2D centered data
np.random.seed(42)
N = 500
X = np.random.randn(N, 2) @ np.array([[2.0, 1.0], [0.5, 1.5]]) # Correlated features
X -= np.mean(X, axis=0) # Center the data

# Compute Covariance Matrix
cov_matrix = (X.T @ X) / N
print("Sample Covariance Matrix:\n", cov_matrix)

# Check Eigenvalues
eigenvals = np.linalg.eigvalsh(cov_matrix)
print("Eigenvalues of Covariance Matrix:", eigenvals)
print("Are all eigenvalues >= 0?", np.all(eigenvals >= 0))
```

## 5. Cholesky Decomposition

### What is Cholesky Decomposition?
Every symmetric Positive Definite matrix $\mathbf{A}$ can be uniquely decomposed into:
$$\mathbf{A} = \mathbf{L} \mathbf{L}^T$$
where $\mathbf{L}$ is a **lower triangular matrix** with strictly positive diagonal entries.

### Why is it essential in AI/ML?
1. **Computational Speed**: Solving $\mathbf{A} \mathbf{x} = \mathbf{b}$ using Cholesky is roughly **twice as fast** and far more numerically stable than standard LU decomposition or matrix inversion.
2. **Sampling Multivariate Gaussians**: To generate samples $\mathbf{y} \sim \mathcal{N}(\mathbf{\mu}, \mathbf{\Sigma})$, we generate standard independent samples $\mathbf{z} \sim \mathcal{N}(\mathbf{0}, \mathbf{I})$ and transform them:
$$\mathbf{y} = \mathbf{\mu} + \mathbf{L} \mathbf{z} \quad 	ext{where } \mathbf{\Sigma} = \mathbf{L} \mathbf{L}^T$$


```python
# 1. Perform Cholesky Decomposition
Sigma = np.array([[4.0, 1.2],
                  [1.2, 2.0]])

# L such that L @ L.T = Sigma
L = np.linalg.cholesky(Sigma)
print("Original Covariance Matrix Sigma:\n", Sigma)
print("Cholesky Factor L (Lower Triangular):\n", L)
print("Reconstruction L @ L.T:\n", L @ L.T)

# 2. Generate 1000 Correlated Gaussian Samples using Cholesky
num_samples = 1000
mu = np.array([2.0, 5.0])
z = np.random.randn(num_samples, 2) # Standard Normal (0, I)
y = mu + z @ L.T                    # Correlated Normal (mu, Sigma)

# Visualize generated samples
plt.figure(figsize=(7, 6))
plt.scatter(y[:, 0], y[:, 1], alpha=0.4, color='teal', label='Samples')
plt.title(r"Samples Generated via Cholesky: $\mathbf{y} = \mathbf{\mu} + \mathbf{L}\mathbf{z}$")
plt.xlabel("X1")
plt.ylabel("X2")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 6. Summary & Key Takeaways

1. **Quadratic forms** $\mathbf{x}^T \mathbf{A} \mathbf{x}$ capture the second-order curvature of multi-variable functions.
2. A matrix is **Positive Definite (PD)** iff all its eigenvalues are strictly positive, guaranteeing a unique global minimum.
3. **Covariance matrices** $\mathbf{\Sigma} = rac{1}{N}\mathbf{X}^T\mathbf{X}$ are naturally positive semi-definite.
4. **Cholesky Factorization** $\mathbf{A} = \mathbf{L}\mathbf{L}^T$ is the most efficient method for solving linear systems and sampling from multivariate Gaussian distributions.
