# 01. Entropy & Information Theory

Welcome to **Information Theory for AI/ML**!

Founded by Claude Shannon in 1948, **Information Theory** is the mathematical framework that quantifies uncertainty, surprise, and information content.
In modern AI/ML, information theory directly defines:
- The loss functions used to train classification models (**Cross-Entropy**).
- The regularization metrics used in generative models like VAEs (**KL Divergence**).
- Decision-tree splitting criteria (**Information Gain**).
- The metric for evaluating Large Language Models (**Perplexity**).

---

## 1. Self-Information (Surprise)

### Definition:
If an event $x$ has probability $P(x)$, its **self-information** (or "surprise") is:

$$I(x) = -\log_2 P(x) = \log_2 \left(\frac{1}{P(x)}\right)$$

- **Units**: If using base 2 ($\log_2$), information is measured in **bits** (or shannons). If using natural log ($\ln$), it is in **nats**.
- **Intuition**:
  - A rare, unlikely event (e.g., "It snowed in the Sahara Desert", $P = 0.0001$) carries **high information**.
  - A certain event (e.g., "The sun rose today", $P = 1.0$) carries **zero surprise** ($I = -\log_2(1) = 0$).


```python
import numpy as np
import matplotlib.pyplot as plt

# Plotting Self-Information vs Probability
p_vals = np.linspace(0.001, 1.0, 300)
info_bits = -np.log2(p_vals)

plt.figure(figsize=(8, 5))
plt.plot(p_vals, info_bits, 'b-', linewidth=2)
plt.title(r"Self-Information: $I(x) = -\log_2 P(x)$")
plt.xlabel("Probability P(x)")
plt.ylabel("Information Content (Bits)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Shannon Entropy (Expected Surprise)

### Definition:
**Shannon Entropy $H(X)$** is the expected (average) information content over all possible outcomes of a discrete random variable $X$:

$$H(X) = \mathbb{E}[I(X)] = -\sum_{i=1}^n P(x_i) \log_2 P(x_i)$$

### Properties of Entropy:
1. **Non-negativity**: $H(X) \geq 0$.
2. **Deterministic Event**: $H(X) = 0$ if one outcome has $P = 1$ and all others $P = 0$.
3. **Maximum Uncertainty**: Entropy is maximized when all outcomes are equally likely (Uniform Distribution):
   $$H_{max} = \log_2(n) \quad \text{for } n \text{ equiprobable states}$$


```python
# Entropy of a Biased Coin Toss
# P(Heads) = p, P(Tails) = 1 - p
p = np.linspace(1e-5, 1 - 1e-5, 300)
entropy = -(p * np.log2(p) + (1 - p) * np.log2(1 - p))

plt.figure(figsize=(8, 5))
plt.plot(p, entropy, 'r-', linewidth=2.5)
plt.axvline(0.5, color='k', linestyle='--', alpha=0.7, label='Max Uncertainty (p=0.5, H=1 bit)')
plt.title("Shannon Entropy of a Binary Random Variable (Coin Toss)")
plt.xlabel("Probability of Heads p")
plt.ylabel("Entropy H(p) in Bits")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Joint & Conditional Entropy

- **Joint Entropy**: Uncertainty of both $X$ and $Y$ occurring together:
  $$H(X, Y) = -\sum_x \sum_y P(x, y) \log_2 P(x, y)$$
- **Conditional Entropy**: Remaining uncertainty in $Y$ given knowledge of $X$:
  $$H(Y|X) = -\sum_x \sum_y P(x, y) \log_2 P(y|x)$$
- **Chain Rule of Entropy**:
  $$H(X, Y) = H(X) + H(Y|X)$$


```python
# Computing Text Entropy of a character sequence
def calculate_text_entropy(text):
    chars, counts = np.unique(list(text), return_counts=True)
    probs = counts / len(text)
    return -np.sum(probs * np.log2(probs))

sample_text = "mathematics for artificial intelligence and machine learning"
ent = calculate_text_entropy(sample_text)
print(f"Text: '{sample_text}'")
print(f"Character-level Shannon Entropy: {ent:.4f} bits/character")
```

---

## 4. Summary & Key Takeaways

1. **Self-Information** $I(x) = -\log_2 P(x)$ quantifies surprise.
2. **Shannon Entropy** $H(X) = -\sum P(x)\log_2 P(x)$ measures the average uncertainty of a distribution.
3. Entropy is **maximized** under a uniform distribution and is **zero** for deterministic outcomes.
4. Entropy is the foundation for Cross-Entropy loss and KL Divergence in modern neural networks.
