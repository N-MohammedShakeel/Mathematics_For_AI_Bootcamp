# 2.1 Random Sampling
Imagine you’re picking names from a hat to form a fair team or tasting a few candies from a mixed bag to guess the flavors—without bias. That’s what random sampling is all about: selecting a subset of data in a way that represents the whole, giving every piece an equal shot. It’s like ensuring your team reflects the class, not just your best friends!

## What Is Random Sampling?
Random sampling involves choosing data points randomly from a larger set to create a manageable subset. Common techniques include:

- **Simple Random Sampling**: Picking items by chance, like drawing names from a hat.
- **Stratified Sampling**: Dividing into groups (e.g., by age) and sampling from each, like ensuring both kids and adults are included.

Let’s use a custom dataset of random data. We’ll generate a dataset with 100 entries and save it as `random_data.csv` in your `data` folder. Load it like this:

```python
import pandas as pd
import numpy as np
import os

# Generate a fair amount of data
os.makedirs("data", exist_ok=True)
n = 100
np.random.seed(42)  # For reproducibility
data = pd.DataFrame({
    "ID": range(1, n + 1),
    "Age": np.random.randint(9, 14, n),  # Ages 9 to 13
    "Score": np.random.randint(70, 100, n)  # Scores 70 to 99
})
data.to_csv('./data/random_data.csv', index=False)
print("Dataset saved as data/random_data.csv with {} entries".format(n))

# Load the dataset
data = pd.read_csv('./data/random_data.csv')

print(data.head())  # Show the first 5 rows
```

Now, let’s perform sampling. For simple random sampling (e.g., 20 out of 100 rows):

```python
# Simple random sampling
sample_size = 20
sample = data.sample(n=sample_size, random_state=42) # sample () is a method of pandas DataFrame to randomly select rows
print("Random Sample ({} rows):".format(sample_size))
print(sample)
```

For stratified sampling by age group (e.g., <11, 11+), use:

```python
# Stratified sampling by age group

# this line groups the data by age >= 11 and then samples 20% from each group
# the group_keys=False argument prevents the group labels from being added to the index of the result
stratified_sample = data.groupby(data['Age'] >= 11, group_keys=False).apply(lambda x: x.sample(frac=0.2))
print("Stratified Sample:")
print(stratified_sample)
```

This ensures both age groups (<11 and 11+) are represented, with about 20% from each group.

## Why Is This Necessary?

- **In Mathematics**: It provides a fair basis for estimating population stats, like average scores.
- **In Machine Learning (ML)**: It ensures training data mirrors the full dataset, avoiding skewed models.

## Relevance in Machine Learning
Random sampling is the backbone of ML data prep. A good sample prevents bias in training (e.g., overrepresenting high scores), while stratified sampling balances rare categories (e.g., older students). It’s key to building reliable predictors.

## Applications

- **Surveys**: Polling a random group to predict election results.
- **A/B Testing**: Testing website versions with random user subsets to measure impact.

## Step-by-Step Example
Let’s sample our data:

1. **Load the Data**: Import `random_data.csv` from `data/`.
2. **Simple Sampling**: Pick 20 random rows to estimate the average score.
3. **Stratified Sampling**: Split by age (e.g., <11, 11+) and sample 20% from each.

Run the code above—your sample might show a mix of scores and ages, giving a fair peek at the data!

## Practical Insights

- **Sample Size**: Larger samples reduce error but take more time—balance is key.
- **Stratification**: Use it when groups (e.g., age) matter to avoid missing minorities.
- **Reproducibility**: Set `random_state` for consistent results during testing.

Let’s calculate the mean score of the simple random sample:

```python
# Calculate mean score of the simple random sample
mean_score = sample['Score'].mean()
print("Mean score of the random sample: {:.2f}".format(mean_score))
```

## Common Pitfalls to Avoid

- **Bias Creep**: Non-random picks (e.g., only high scores) skew results—use proper randomization.
- **Small Samples**: Too few rows (e.g., 1) miss the big picture—aim for at least 10-20% of data.
- **Over-Sampling**: Repeatedly sampling the same data distorts estimates—ensure fresh draws.

Let’s compare the stratified sample’s mean score:

```python
# Calculate mean score of the stratified sample
mean_score_stratified = stratified_sample['Score'].mean()
print("Mean score of the stratified sample: {:.2f}".format(mean_score_stratified))
```
