# 2.4 Resampling and Bootstrap
Imagine reshuffling a deck of cards to deal new hands or re-tasting a small candy batch to double-check the flavor—without needing more data. That’s what resampling and bootstrap are about: reusing your existing sample to estimate stats and uncertainties, like getting extra practice from the same set of cards. It’s a clever way to make the most of what you’ve got!

## What Are Resampling and Bootstrap?

- **Resampling**: Reusing data in different ways, such as drawing samples with replacement.
- **Bootstrap**: A specific resampling method where you repeatedly sample with replacement from your data to estimate things like means or confidence intervals.

Let’s simulate this with Python. We’ll generate a custom dataset of scores and apply bootstrap resampling:

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a small sample dataset (e.g., 10 scores)
np.random.seed(42)
original_data = np.random.normal(loc=70, scale=10, size=10).round()
print("Original Data:", original_data)

# Bootstrap resampling: 1000 samples with replacement
n_bootstraps = 1000
bootstrap_means = np.array([np.mean(np.random.choice(original_data, size=len(original_data), replace=True)) 
                           for _ in range(n_bootstraps)])

# Plot the distribution of bootstrap means
plt.hist(bootstrap_means, bins=30, edgecolor='black', density=True)
plt.title('Bootstrap Distribution of Means')
plt.xlabel('Mean')
plt.ylabel('Density')
plt.show()

# Estimate 95% confidence interval
confidence_interval = np.percentile(bootstrap_means, [2.5, 97.5])
print("95% Confidence Interval: [{:.2f}, {:.2f}]".format(confidence_interval[0], confidence_interval[1]))
```

This code generates 10 scores (around a mean of 70), creates 1,000 bootstrap samples, and plots their means. The confidence interval (e.g., [65.23, 74.87]) shows where the true mean likely lies.

## Why Is This Necessary?

- **In Mathematics**: It boosts reliability when you can’t collect more data, like refining estimates from a single survey.
- **In Machine Learning (ML)**: It quantifies uncertainty in model predictions, improving trust in results.

## Relevance in Machine Learning
Bootstrap is ML’s safety net. It helps estimate how much models might vary (e.g., predicting test scores), supports confidence intervals for feature importance, and validates models when data is scarce. It’s essential for robust learning.

## Applications

- **Confidence Intervals**: Pinpointing the range for a population mean from a small sample.
- **Model Validation**: Testing model stability across resampled datasets.

## Step-by-Step Example
Let’s bootstrap our data:

1. **Generate Data**: Create 10 random scores with `np.random.normal`.
2. **Resample with Replacement**: Draw 1,000 bootstrap samples and compute means.
3. **Analyze Results**: The histogram shows a bell curve, and the confidence interval estimates the true mean.

Run the code above—the plot and interval (e.g., [65.23, 74.87]) give a solid estimate!

## Practical Insights

- **Sample Size**: Even small datasets (e.g., 10) work with enough bootstraps (e.g., 1,000).
- **Replacement Key**: Sampling with replacement mimics new data—don’t skip it.
- **Iteration Count**: More bootstraps (e.g., 10,000) refine intervals—start with 1,000.

Let’s try with more bootstraps (e.g., 10,000) to see the effect:

```python
# Bootstrap with 10,000 samples
n_bootstraps_large = 10000
bootstrap_means_large = np.array([np.mean(np.random.choice(original_data, size=len(original_data), replace=True)) 
                                for _ in range(n_bootstraps_large)])

# Plot the distribution of larger bootstrap means
plt.hist(bootstrap_means_large, bins=30, edgecolor='black', density=True)
plt.title('Bootstrap Distribution of Means (10,000 Samples)')
plt.xlabel('Mean')
plt.ylabel('Density')
plt.show()

# Estimate 95% confidence interval with more samples
confidence_interval_large = np.percentile(bootstrap_means_large, [2.5, 97.5])
print("95% Confidence Interval (10,000 samples): [{:.2f}, {:.2f}]".format(confidence_interval_large[0], confidence_interval_large[1]))
```

## Common Pitfalls to Avoid

- **No Replacement**: Sampling without replacement defeats bootstrap—always use replacement.
- **Too Few Bootstraps**: 10 samples won’t stabilize the distribution—aim for 1,000+.
- **Biased Original Data**: If the initial sample is skewed, bootstrap won’t fix it—ensure fair sampling first.

Let’s calculate the standard error from the bootstrap means for a measure of uncertainty:

```python
# Calculate standard error from bootstrap means
bootstrap_std = np.std(bootstrap_means_large)
print("Standard Error from Bootstrap Means: {:.2f}".format(bootstrap_std))
```
