# 2.2 Sample Mean and Variance
Imagine you’re guessing the average height of your class by measuring just a few friends, or checking if a candy bag has consistent sweetness by tasting a handful. That’s what sample mean and variance are about—using a small chunk of data to estimate the average and spread of a whole population. It’s like getting a sneak peek at the big picture with limited clues!

## What Are Sample Mean and Variance?

- **Sample Mean**: The average of your sample, like adding up a few heights and dividing by the count to guess the class average.
- **Sample Variance**: How much the sample values differ from the mean, showing if your candy tastes vary widely or stay similar.

Let’s use a custom dataset of sampled data. We’ll generate a dataset with 50 entries and save it as `sampled_data.csv` in your `data` folder. Load it like this:

```python
import pandas as pd
import numpy as np
import os

# Generate a fair amount of data
os.makedirs("data", exist_ok=True)
n = 50
np.random.seed(42)  # For reproducibility
data = pd.DataFrame({
    "ID": range(1, n + 1),
    "Score": np.random.normal(85, 5, n).round()  # Mean 85, std 5
})
data.to_csv('data/sampled_data.csv', index=False)
print("Dataset saved as data/sampled_data.csv with {} entries".format(n))

# Load the dataset
data = pd.read_csv('data/sampled_data.csv')

print(data.head())  # Show the first 5 rows
```

Now, let’s calculate the sample mean and variance:

```python
# Calculate sample mean
sample_mean = data['Score'].mean()
print("Sample Mean: {:.2f}".format(sample_mean))

# Calculate sample variance
sample_variance = data['Score'].var(ddof=1)  # ddof=1 for sample variance
print("Sample Variance: {:.2f}".format(sample_variance))
```

This code gives a mean around 85 (based on the generated data) and a variance reflecting the spread, showing the average score and how much scores vary.

## Why Is This Necessary?

- **In Mathematics**: It lets us estimate population stats without measuring everyone, like predicting class height from a few students.
- **In Machine Learning (ML)**: It helps evaluate model performance on samples, guiding training adjustments.

## Relevance in Machine Learning
Sample mean and variance are ML’s reality check. The mean sets a baseline for predictions (e.g., average score), while variance flags data inconsistency that might confuse models. They’re crucial for testing how well a model generalizes.

## Applications

- **Population Studies**: Estimating average income from a survey sample.
- **Quality Control**: Checking product weight variance in a factory batch.

## Step-by-Step Example
Let’s estimate from our data:

1. **Load the Data**: Import `sampled_data.csv` from `data/`.
2. **Compute Mean**: Average the scores to get around 85—our estimated class score!
3. **Calculate Variance**: Measure spread from the mean, giving a value like 25, showing score variation.

Run the code above to confirm—our sample hints at a stable score range!

## Practical Insights

- **Sample Size**: Bigger samples (e.g., 30+) give better estimates—small ones like 3 are rough guesses.
- **Bias Check**: Ensure the sample mirrors the population to avoid skewed means.
- **Variance Use**: High variance signals outliers—dig deeper if needed.

Let’s visualize the data with a simple plot to see the spread:

```python
import matplotlib.pyplot as plt

# Plot scores
plt.hist(data['Score'], bins=10, edgecolor='black')
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.title('Score Distribution')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

## Common Pitfalls to Avoid

- **Small Sample Error**: Three scores might miss the true mean—gather more data if possible (we used 50 here).
- **Population Mismatch**: A sample of only high scores skews the mean—randomize selection.
- **Variance Misinterpretation**: A low variance doesn’t mean accuracy—check context.

Let’s calculate the standard deviation for a better spread insight:

```python
# Calculate sample standard deviation
sample_std = data['Score'].std(ddof=1)
print("Sample Standard Deviation: {:.2f}".format(sample_std))
```
