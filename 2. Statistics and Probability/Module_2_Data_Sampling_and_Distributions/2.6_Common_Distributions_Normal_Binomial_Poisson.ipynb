# 2.6 Common Distributions: Normal, Binomial, Poisson
Imagine measuring the heights of people in a town, flipping a coin many times to track heads, or counting customers arriving at a store each hour. These scenarios follow key probability distributions—Normal, Binomial, and Poisson—that help us model and predict data behavior. Think of them as blueprints for understanding randomness, whether it’s a bell curve of heights, a tally of wins, or a stream of arrivals. Let’s dive deep into each with visualizations to bring them to life!

## What Are These Distributions?

- **Normal Distribution**: A bell-shaped curve where data clusters around the mean, like heights or test scores. It’s defined by mean (μ) and standard deviation (σ).
- **Binomial Distribution**: Models the number of successes in a fixed number of trials (e.g., heads in 10 coin flips), with parameters trials (n) and success probability (p).
- **Poisson Distribution**: Counts events in a fixed interval (e.g., customer arrivals per hour), with a single parameter, the average rate (λ).

Let’s generate and visualize these distributions with Python:

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)

# 1. Normal Distribution (e.g., heights, mean=170 cm, std=10 cm, 1000 samples)
normal_data = np.random.normal(loc=170, scale=10, size=1000)
plt.figure(figsize=(12, 4))
sns.histplot(normal_data, bins=30, kde=True, color='blue', stat='density')
plt.axvline(np.mean(normal_data), color='red', linestyle='--', label=f'Mean = {np.mean(normal_data):.1f} cm')
plt.axvline(np.mean(normal_data) - np.std(normal_data), color='green', linestyle=':', label='±1σ')
plt.axvline(np.mean(normal_data) + np.std(normal_data), color='green', linestyle=':')
plt.title('Normal Distribution (Heights, μ=170, σ=10)')
plt.xlabel('Height (cm)')
plt.ylabel('Density')
plt.legend()
plt.show()

# 2. Binomial Distribution (e.g., 10 coin flips, p=0.5, 1000 trials)
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)
plt.figure(figsize=(12, 4))
sns.histplot(binomial_data, bins=11, color='green', stat='density', discrete=True)
plt.axvline(np.mean(binomial_data), color='red', linestyle='--', label=f'Mean = {np.mean(binomial_data):.1f}')
plt.title('Binomial Distribution (10 Flips, p=0.5)')
plt.xlabel('Number of Heads')
plt.ylabel('Density')
plt.legend()
plt.show()

# 3. Poisson Distribution (e.g., 3 customers/hour, 1000 intervals)
poisson_data = np.random.poisson(lam=3, size=1000)
plt.figure(figsize=(12, 4))
sns.histplot(poisson_data, bins=15, color='red', stat='density', discrete=True)
plt.axvline(np.mean(poisson_data), color='red', linestyle='--', label=f'Mean = {np.mean(poisson_data):.1f}')
plt.title('Poisson Distribution (λ=3 Customers/Hour)')
plt.xlabel('Number of Customers')
plt.ylabel('Density')
plt.legend()
plt.show()
```

These plots show:

- **Normal**: A smooth bell curve centered around the mean (approximately 170 cm), with 68% of data within ±1 standard deviation.
- **Binomial**: A bar chart peaking around 5 heads (expected with p=0.5), reflecting the binomial nature of binary outcomes.
- **Poisson**: A skewed bar chart clustering around 3 customers, typical for rare event counts.

## In-Depth Explanation with Visualizations

### Normal Distribution
The bell shape arises because most data points (e.g., approximately 68% within 1 standard deviation, 95% within 2 standard deviations) cluster near the mean, following the empirical rule. It’s ideal for continuous data like heights or IQ scores. The visualization shows symmetry and tapering tails, with the mean and ±1σ lines highlighting this pattern. Mathematically, the probability density function is ( f(x) = rac{1}{sigma sqrt{2pi}} e^{-rac{(x-mu)^2}{2sigma^2}} ), central to statistical inference and ML error modeling.

### Binomial Distribution
This discrete distribution models the number of successes in ( n ) independent trials, each with success probability ( p ). For 10 flips with ( p=0.5 ), the peak at 5 reflects the equal chance, with probabilities dropping for extremes (e.g., 0 or 10 heads). The probability mass function is ( P(k) = inom{n}{k} p^k (1-p)^{n-k} ), where ( inom{n}{k} ) is the binomial coefficient. The bar chart highlights this stair-step pattern, useful for classification tasks in ML.

### Poisson Distribution
Used for rare events over a fixed interval, it’s skewed right for low ( lambda ) (e.g., 3). As ( lambda ) increases (e.g., to 20), it approximates a Normal distribution due to the Central Limit Theorem. The probability mass function is ( P(k) = rac{lambda^k e^{-lambda}}{k!} ), and the plot shows a peak at 3, with fewer occurrences at higher counts. This is critical for event prediction in ML, like network traffic analysis.

Let’s visualize the Poisson distribution with a higher ( lambda ) (e.g., 20) to see the Normal approximation:

```python
# Poisson Distribution with higher λ (e.g., 20 customers/hour, 1000 intervals)
poisson_data_high = np.random.poisson(lam=20, size=1000)
plt.figure(figsize=(12, 4))
sns.histplot(poisson_data_high, bins=30, color='purple', stat='density', discrete=True)
plt.axvline(np.mean(poisson_data_high), color='red', linestyle='--', label=f'Mean = {np.mean(poisson_data_high):.1f}')
plt.title('Poisson Distribution (λ=20 Customers/Hour)')
plt.xlabel('Number of Customers')
plt.ylabel('Density')
plt.legend()
plt.show()
```

## Practical Insights

- **Parameter Tuning**: Adjust ( mu ), ( sigma ) (Normal), ( n ), ( p ) (Binomial), or ( lambda ) (Poisson) to fit your data—test with plots.
- **Sample Size**: Larger samples (e.g., 1,000) smooth histograms—start with 100+.
- **Real-World Fit**: Check data against these distributions (e.g., use Q-Q plots or statistical tests like Kolmogorov-Smirnov) to ensure applicability.

## Common Pitfalls to Avoid

- **Misapplying Distributions**: Using Normal for count data (e.g., customers) instead of Poisson leads to errors—match the data type.
- **Small Samples**: With 10 points, distributions look jagged—use 30+ for clarity.
- **Ignoring Assumptions**: Binomial assumes independent trials; violating this (e.g., biased coins) skews results. Poisson assumes events are rare and independent.
