# 2.5 Confidence Intervals
Imagine guessing how fast a runner will finish a race based on a few practice times, or polling a crowd to estimate how many like pizza with a range of uncertainty. That’s what confidence intervals do— they give a range where a population parameter (like a mean) likely falls, adding a safety net to your estimates. It’s like saying, “I’m 95% sure the finish time is between 10 and 12 minutes!”

## What Are Confidence Intervals?
A confidence interval is a range, calculated from a sample, that likely contains the true population parameter (e.g., mean) with a set confidence level (e.g., 95%). It uses the sample mean and standard error, often assuming a normal distribution from the Central Limit Theorem.

Let’s simulate this with Python. We’ll generate a custom dataset of scores and compute a confidence interval:

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Simulate a sample dataset (e.g., 30 scores)
np.random.seed(42)
sample_data = np.random.normal(loc=70, scale=10, size=30).round()

# Calculate sample mean and standard error
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1) 
standard_error = sample_std / np.sqrt(len(sample_data))

# Calculate 95% confidence interval

# stats.t.ppf is used to get the critical t-value for the given confidence level and degrees of freedom
# the t-value means how many standard errors we need to go out from the mean to capture the desired confidence level
confidence_level = 0.95
t_critical = stats.t.ppf((1 + confidence_level) / 2, df=len(sample_data)-1) 
margin_of_error = t_critical * standard_error
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)


print(f"Sample Mean: {sample_mean:.2f}")
print(f"Standard Error: {standard_error:.2f}")
print(f"95% Confidence Interval: [{confidence_interval[0]:.2f}, {confidence_interval[1]:.2f}]")

# Plot sample data with CI
plt.hist(sample_data, bins=15, edgecolor='black', density=True)
plt.axvline(confidence_interval[0], color='r', linestyle='--', label='CI Lower')
plt.axvline(confidence_interval[1], color='r', linestyle='--', label='CI Upper')
plt.axvline(sample_mean, color='g', linestyle='-', label='Mean')
plt.title('Sample Data with 95% Confidence Interval')
plt.xlabel('Score')
plt.ylabel('Density')
plt.legend()
plt.show()
```

This code generates 30 scores (around a mean of 70), computes a 95% confidence interval (e.g., [66.12, 73.88]), and plots it, showing where the true mean likely sits.

## Why Is This Necessary?

- **In Mathematics**: It quantifies how precise our sample estimate is, like narrowing down a poll result.
- **In Machine Learning (ML)**: It assesses model robustness, ensuring predictions aren’t flukes.

## Relevance in Machine Learning
Confidence intervals are ML’s trust meter. They show how stable a model’s output (e.g., mean prediction) is across data, help compare models, and guide decisions in A/B testing. They’re vital for reliable deployment.

## Applications

- **Survey Results**: Estimating the true support for a candidate with a 95% confidence range.
- **A/B Testing**: Determining if a new website layout improves clicks within a confidence interval.

## Step-by-Step Example
Let’s build a confidence interval:

1. **Simulate Data**: Generate 30 random scores with `np.random.normal`.
2. **Calculate Mean and Error**: Find the average and standard error.
3. **Determine Interval**: Use a t-score for a 95% range (e.g., [66.12, 73.88]).

Run the code above—the plot and interval confirm where the true mean likely hides!

## Practical Insights

- **Sample Size**: Larger samples (e.g., 50+) tighten the interval—30 is a good start.
- **Confidence Level**: 95% is common, but 99% widens the range—choose based on need.
- **Normality**: The CLT helps, but very small samples need caution.

Let’s try a 99% confidence interval for a wider range:

```python
# Calculate 99% confidence interval
confidence_level_99 = 0.99
t_critical_99 = stats.t.ppf((1 + confidence_level_99) / 2, df=len(sample_data)-1)
margin_of_error_99 = t_critical_99 * standard_error
confidence_interval_99 = (sample_mean - margin_of_error_99, sample_mean + margin_of_error_99)

print(f"99% Confidence Interval: [{confidence_interval_99[0]:.2f}, {confidence_interval_99[1]:.2f}]")

# Plot with 99% CI
plt.hist(sample_data, bins=15, edgecolor='black', density=True)
plt.axvline(confidence_interval_99[0], color='r', linestyle='--', label='99% CI Lower')
plt.axvline(confidence_interval_99[1], color='r', linestyle='--', label='99% CI Upper')
plt.axvline(sample_mean, color='g', linestyle='-', label='Mean')
plt.title('Sample Data with 99% Confidence Interval')
plt.xlabel('Score')
plt.ylabel('Density')
plt.legend()
plt.show()
```

## Common Pitfalls to Avoid

- **Small Samples**: With 5 scores, the interval might be unreliable—use 30+.
- **Misinterpreting**: A 95% interval doesn’t mean there’s a 95% chance the mean is inside— it’s about repeated sampling.
- **Ignoring Variability**: If data varies wildly, the interval grows—check the standard deviation.

Let’s check the sample standard deviation to assess variability:

```python
# Calculate sample standard deviation
print(f"Sample Standard Deviation: {sample_std:.2f}")
```
