# 2.7 Bayes Theorem
Imagine you're a detective solving a mystery, starting with a hunch and updating it as new clues come in—like guessing if it's raining based on wet streets, then adjusting when you see umbrellas. That's what Bayes Theorem is all about: using conditional probability to update your beliefs with evidence. It's like refining a guess as more facts roll in, turning uncertainty into smarter decisions!

## What Is Bayes’ Theorem?

Bayes’ Theorem is a simple formula to find the probability of something happening given new information.
It’s based on **conditional probability**, which is the chance of one event if another is true — like the chance of a home team winning if it's raining.

The formula is:

```math
P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}
```

* **P(A|B)** — Posterior probability: probability of A given B (updated belief after seeing evidence).
* **P(B|A)** — Likelihood: how likely the evidence is if A is true.
* **P(A)** — Prior probability: your starting belief before new evidence.
* **P(B)** — Evidence: the total probability of seeing the new information.

---

### Related Concepts

* **Conditional Probability:**

  ```math
  P(A|B) = \frac{P(A \text{ and } B)}{P(B)}
  ```

  It’s the foundation of Bayes’ Theorem, showing how one event depends on another.

* **Prior and Posterior:**
  The prior is your initial guess (e.g., “20% chance of spam”), and the posterior is the updated guess after evidence (e.g., “60% chance if it has a link”).

* **Joint Probability:**

  ```math
  P(A \text{ and } B)
  ```

  The probability that both events happen — ties into the likelihood and evidence.

* **Marginal Probability:**

  ```math
  P(B)
  ```

  Calculated by summing over all possibilities, it normalizes the equation.

Let’s use a custom binary dataset for spam detection.
Run the script `script_to_generate_spam_data.py` first to create `spam_data.csv` in the `data` folder with this data:

| Email ID | Has Link (1=Yes) | Is Spam (1=Yes) |
| -------- | ---------------- | --------------- |
| 1        | 1                | 1               |
| 2        | 0                | 0               |
| 3        | 1                | 1               |
| 4        | 0                | 0               |
| 5        | 1                | 0               |

Now, let’s calculate with Python:




````python
import pandas as pd

# Import the custom binary dataset from the data folder
data = pd.read_csv('data/spam_data.csv')

# Calculate conditional probabilities
p_spam = data['Is Spam (1=Yes)'].mean()  # Prior P(Spam) = 0.4 , 
p_link_given_spam = data[data['Is Spam (1=Yes)'] == 1] ['Has Link (1=Yes)'].mean()  # P(Link|Spam) = 1.0  
p_link = data['Has Link (1=Yes)'].mean()  # P(Link) = 0.6

# Bayes Theorem: P(Spam|Link) = P(Link|Spam) * P(Spam) / P(Link)
p_spam_given_link = (p_link_given_spam * p_spam) / p_link

print(f"Prior Probability of Spam: {p_spam:.2f}")
print(f"Likelihood P(Link|Spam): {p_link_given_spam:.2f}")
print(f"Evidence P(Link): {p_link:.2f}")
print(f"Posterior Probability of Spam given Link: {p_spam_given_link:.2f}")
````

This calculates the probability an email is spam if it has a link—about 67%! With 2 out of 5 emails as spam (40%), and all spam emails having links (100%), the presence of a link raises the spam likelihood from 40% to 67%.

2. **Calculate Prior**

   ```math
   P(\text{Spam}) = \frac{2}{5} = 0.4

   ```

   Substitute the values:

   ```math
   P(\text{Spam}|\text{Link}) = \frac{1.0 \times 0.4}{0.6} = 0.67
   ```

## In-Depth Explanation with Related Concepts

### How It Works

Bayes’ Theorem flips conditional probability.
Instead of asking “What’s the chance of rain if it’s cloudy?” (likelihood), it asks “What’s the chance of rain given wet streets?” (posterior).
It uses the **prior** (initial rain chance), **likelihood** (wet streets if raining), and **evidence** (overall wet street chance) to compute this.

---

### Law of Total Probability

The evidence ( P(B) ) is found by considering all ways ( B ) can happen:

$$
P(B) = P(B|A) \cdot P(A) + P(B|\neg A) \cdot P(\neg A)
$$

For our spam example:

$$
P(\text{Link}) = P(\text{Link}|\text{Spam}) \cdot P(\text{Spam}) + P(\text{Link}|\neg \text{Spam}) \cdot P(\neg \text{Spam})
$$

where
$$
P(\neg \text{Spam}) = 0.6
\quad \text{and} \quad
P(\text{Link}|\neg \text{Spam}) = \frac{1}{3}
$$
(1 link in 3 non-spam emails).

---

### Bayesian Updating

You can chain updates.
If a second clue (e.g., “has attachment”) appears, use the new posterior as the prior and repeat.
This iterative process is the heart of **Bayesian learning** in machine learning.

Let’s extend our example with a second feature (has attachment):


````python
# Add attachment data (hypothetical extension)
data['Has Attachment (1=Yes)'] = [1, 0, 1, 0, 0]  # Example data
p_attachment_given_spam = data[data['Is Spam (1=Yes)'] == 1]['Has Attachment (1=Yes)'].mean()  # P(Attachment|Spam) = 1.0
p_attachment = data['Has Attachment (1=Yes)'].mean()  # P(Attachment) = 0.4

# New posterior: P(Spam|Link and Attachment)
# Assuming independence for simplicity: P(Link and Attachment) ≈ P(Link) * P(Attachment)
p_spam_given_link_and_attachment = (p_link_given_spam * p_attachment_given_spam * p_spam) / (p_link * p_attachment)
print(f"Posterior Probability of Spam given Link and Attachment: {p_spam_given_link_and_attachment:.2f}")
````

This shows how multiple clues (link and attachment) can refine the probability further (e.g., to 100% with this small dataset, assuming independence).

### Intuitive Analogy
Think of it as a recipe: Start with a base belief (prior), add evidence (likelihood), and divide by the total evidence mix (marginal) to get a new belief (posterior). It’s like adjusting a cake batter as you taste and tweak!

## Practical Insights

- **Priors Matter**: A strong prior (e.g., 90% spam) heavily influences updates—choose based on domain knowledge.
- **Evidence Quality**: Weak evidence (e.g., common links) won’t change much—seek distinctive features.
- **Chaining Updates**: Apply Bayes repeatedly as new data arrives for ongoing refinement, ideal for dynamic ML models.

Let’s simulate a larger dataset to see stability:

````python
import numpy as np

# Simulate larger dataset (100 emails)
np.random.seed(42)
n_samples = 100
has_link = np.random.binomial(1, 0.6, n_samples)
is_spam = np.random.binomial(1, 0.4, n_samples)  # True spam rate
larger_data = pd.DataFrame({'Has Link (1=Yes)': has_link, 'Is Spam (1=Yes)': is_spam})

# Recalculate with larger data
p_spam_large = larger_data['Is Spam (1=Yes)'].mean()
p_link_given_spam_large = larger_data[larger_data['Is Spam (1=Yes)'] == 1]['Has Link (1=Yes)'].mean()
p_link_large = larger_data['Has Link (1=Yes)'].mean()
p_spam_given_link_large = (p_link_given_spam_large * p_spam_large) / p_link_large

print(f"Large Dataset - Prior P(Spam): {p_spam_large:.2f}")
print(f"Large Dataset - P(Spam|Link): {p_spam_given_link_large:.2f}")
````

## Common Pitfalls to Avoid

- **Bad Priors**: A biased starting belief (e.g., assuming all emails are spam) skews results—use neutral or data-driven priors.
- **Ignoring Evidence**: Overlooking ( P(B) ) inflates probabilities—always compute it from all data.
- **Small Samples**: With 5 emails, estimates are rough—use larger datasets (e.g., 100+) for accuracy.

Let’s verify ( P(B) ) using the law of total probability:

````python
# Verify P(Link) with law of total probability
p_link_not_spam = data[data['Is Spam (1=Yes)'] == 0]['Has Link (1=Yes)'].mean()
p_not_spam = 1 - p_spam
p_link_total = (p_link_given_spam * p_spam) + (p_link_not_spam * p_not_spam)
print(f"Verified P(Link) via Total Probability: {p_link_total:.2f}")
````
