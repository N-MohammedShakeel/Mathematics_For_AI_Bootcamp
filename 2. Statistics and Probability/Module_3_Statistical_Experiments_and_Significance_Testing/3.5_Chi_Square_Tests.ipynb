# 3.5 Chi-Square Tests
Imagine you’re piecing together a puzzle to see if two patterns match, or analyzing if voters’ preferences align with their age groups—that’s what chi-square tests are for! They check if two categorical variables are independent or related, helping you uncover hidden connections. It’s like being a detective spotting clues in a crowd!

## What Are Chi-Square Tests?
A chi-square test assesses the relationship between two categorical variables (e.g., gender and survival). It determines if the variables are independent or if there’s a significant association. The key type is:

- **Chi-Square Test of Independence**: Tests if variables are unrelated (e.g., are survival rates the same across genders?).

It uses a **chi-square statistic** (sum of squared differences between observed and expected counts) and a **p-value**. A low p-value (e.g., <0.05) rejects the null hypothesis (H₀) of independence.

We’ll use the Kaggle dataset “Titanic: Machine Learning from Disaster” to test if survival depends on gender. Download the `train.csv` file from: https://www.kaggle.com/c/titanic/data. Place it in the `data` folder.

Let’s analyze it with Python:

````python
import pandas as pd
import scipy.stats as stats

# Load the Kaggle dataset (download and place in data/ folder)
data = pd.read_csv('data/train.csv')

# Drop rows with missing values in 'Sex' or 'Survived'
data = data.dropna(subset=['Sex', 'Survived'])

# Create a contingency table for survival vs. gender
contingency_table = pd.crosstab(data['Sex'], data['Survived'])
print("Contingency Table:")
print(contingency_table)

# Perform chi-square test of independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

# Print results
print(f"Chi-Square Statistic: {chi2_stat:.2f}")
print(f"P-Value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H₀: Survival and gender are not independent (related).")
else:
    print("Fail to reject H₀: Survival and gender are independent.")
````

This code builds a contingency table (e.g., male/female vs. survived/died) and runs a chi-square test, typically yielding a p-value near 0.0000, indicating a significant relationship between gender and survival. The dataset has over 700 valid entries after cleaning.

## Why Is This Necessary?

## In-Depth Explanation with Related Concepts

### How It Works
The chi-square test compares observed frequencies in a contingency table to expected frequencies under independence. The statistic is calculated as the sum of (observed - expected)² / expected across all cells. A high value and low p-value indicate dependence.

### Key Concepts

- **Degrees of Freedom (df)**: (rows - 1) * (columns - 1), e.g., (2-1) * (2-1) = 1 for this 2x2 table.
- **Expected Frequencies**: Calculated assuming independence—check if all >5 for validity.
- **Effect Size**: Measures strength of association (e.g., Cramer’s V)—let’s compute it:

```python
# Calculate Cramer’s V for effect size
n = contingency_table.sum().sum()
phi = chi2_stat / n
min_dim = min(contingency_table.shape) - 1
cramers_v = np.sqrt(phi / min_dim)
print(f"Cramer’s V (Effect Size): {cramers_v:.3f}")
```

- **Assumptions**: No expected counts below 5; large sample size improves reliability.

A moderate Cramer’s V (e.g., 0.3) suggests a noticeable but not overwhelming relationship.

### Intuitive Analogy
Think of it as a detective matching puzzle pieces. The chi-square statistic is how well the pieces fit; a low p-value means the patterns align, revealing a clue!

## Practical Insights

- **Sample Size**: Each cell should have at least 5 counts—verify with the contingency table.
- **Expected Frequencies**: Low values (<5) can skew results—ensure sufficient data (over 700 entries here).
- **Multiple Tests**: Adjust alpha (e.g., 0.05/2 = 0.025) if testing multiple pairs to control error.

Let’s simulate a smaller dataset to see chi-square sensitivity:

````python
# Simulate smaller contingency table
small_table = pd.DataFrame([[10, 5], [5, 10]], index=['Male', 'Female'], columns=[0, 1])
chi2_stat_small, p_value_small, dof_small, expected_small = stats.chi2_contingency(small_table)

print("Small Contingency Table:")
print(small_table)
print(f"Small Chi-Square Statistic: {chi2_stat_small:.2f}")
print(f"Small P-Value: {p_value_small:.4f}")
if p_value_small < alpha:
    print("Reject H₀: Significant relationship even with small data.")
else:
    print("Fail to reject H₀: No significant relationship.")
````

## Common Pitfalls to Avoid

- **Small Samples**: With few observations per cell (e.g., <5), results are unreliable—use 20+ per category.
- **Ignoring Assumptions**: Chi-square assumes no zero counts—fill or drop empty cells.
- **Over-Interpretation**: Significance doesn’t mean strong correlation—assess with Cramer’s V or similar.
