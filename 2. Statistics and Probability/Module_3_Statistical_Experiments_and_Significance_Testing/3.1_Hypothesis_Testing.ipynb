# 3.1 Hypothesis Testing
Imagine you’re testing a new recipe to see if it’s tastier than your old one, or trying a new game strategy to check if it wins more matches. That’s what hypothesis testing is like—a way to validate assumptions about data using statistical evidence. It’s like being a scientist in your kitchen or a coach on the field, deciding if a change makes a real difference!

## What Is Hypothesis Testing?
Hypothesis testing checks a claim (hypothesis) about a population using sample data. It’s a structured way to decide if your hunch holds up with evidence. Key components include:

- **Null Hypothesis (H₀)**: The default assumption, e.g., “the new recipe isn’t better” or “there’s no difference in costs.”
- **Alternative Hypothesis (H₁)**: The claim you’re testing, e.g., “the new recipe is better” or “smokers have higher costs.”
- **P-value**: The probability of seeing your data (or more extreme) if H₀ is true. A low value (e.g., <0.05) suggests rejecting H₀.
- **Significance Level (α)**: A threshold (e.g., 0.05) to decide if the p-value is small enough to reject H₀.

We’ll use the Kaggle dataset “Medical Cost Personal Datasets” for this example. Download it from: https://www.kaggle.com/datasets/mirichoi0218/insurance. This dataset contains medical costs with features like age, BMI, and charges, including a `smoker` column (yes/no). Place the file `insurance.csv` in the `data` folder.

Let’s test if smokers have higher medical costs using a two-sample t-test:

```python
import pandas as pd
import scipy.stats as stats

# Load the Kaggle dataset (download and place in data/ folder)
data = pd.read_csv('data/insurance.csv')

# Separate data into smokers and non-smokers
smokers = data[data['smoker'] == 'yes']['charges']
non_smokers = data[data['smoker'] == 'no']['charges']

# Perform a two-sample t-test (assuming unequal variances)
t_stat, p_value = stats.ttest_ind(smokers, non_smokers, equal_var=False)

# Print results
print(f"Sample Size - Smokers: {len(smokers)}, Non-Smokers: {len(non_smokers)}")
print(f"T-Statistic: {t_stat:.2f}")
print(f"P-Value: {p_value:.4f}")

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H₀: Smokers have significantly higher medical costs.")
else:
    print("Fail to reject H₀: No significant difference in costs.")
```

This code runs a t-test, typically yielding a p-value close to 0.0000, indicating smokers’ medical costs are significantly higher than non-smokers’. The dataset has over 1,300 rows, providing a robust sample.

## In-Depth Explanation with Related Concepts

### How It Works
Hypothesis testing is like a courtroom trial. H₀ is the innocent-until-proven-guilty stance, and the p-value is the evidence strength. If the p-value is low (below α), the jury (you) rejects H₀ in favor of H₁. The t-test compares means, assuming data follows a normal distribution (large samples help via the Central Limit Theorem).

### Key Concepts

- **Type I Error (α)**: Rejecting H₀ when it’s true (false positive), e.g., saying a drug works when it doesn’t.
- **Type II Error (β)**: Failing to reject H₀ when it’s false (false negative), e.g., missing a working drug.
- **Power (1 - β)**: The chance of correctly rejecting H₀ when it’s false, boosted by larger samples.
- **Confidence Interval**: A range (e.g., 95%) where the true mean likely lies, complementing the p-value.

Let’s calculate a confidence interval for the cost difference:

```python
import numpy as np

# Calculate 95% confidence interval for the difference in means
confidence_level = 0.95
mean_diff = smokers.mean() - non_smokers.mean()
std_err = np.sqrt(smokers.var(ddof=1) / len(smokers) + non_smokers.var(ddof=1) / len(non_smokers))
margin_of_error = stats.t.ppf((1 + confidence_level) / 2, df=len(smokers) + len(non_smokers) - 2) * std_err
ci_lower = mean_diff - margin_of_error
ci_upper = mean_diff + margin_of_error

print(f"Mean Difference in Costs (Smokers - Non-Smokers): ${mean_diff:.2f}")
print(f"95% Confidence Interval: (${ci_lower:.2f}, ${ci_upper:.2f})")
```

This interval (e.g., $10,000 to $12,000) shows the cost difference is positive and significant, aligning with the t-test result.

### Intuitive Analogy
Think of it as a taste test. H₀ is “no taste difference,” and the p-value is how surprised you’d be by the results if H₀ were true. A low p-value means the new recipe’s taste stands out!

## Practical Insights

- **Sample Size**: Larger samples (e.g., 1,000+ rows in this dataset) strengthen results—check with `len()`.
- **Significance Level**: 0.05 is standard, but 0.01 is stricter for high-stakes decisions—adjust based on context.
- **Test Choice**: Use t-tests for means (like costs), chi-square for categorical data (like pass/fail), or ANOVA for multiple groups—match the data type.

Let’s explore sample size impact with a subset:

```python
# Subset to 50 samples each for demonstration
smokers_subset = smokers.sample(n=50, random_state=42)
non_smokers_subset = non_smokers.sample(n=50, random_state=42)
t_stat_subset, p_value_subset = stats.ttest_ind(smokers_subset, non_smokers_subset, equal_var=False)

print(f"Subset (50 each) - T-Statistic: {t_stat_subset:.2f}")
print(f"Subset (50 each) - P-Value: {p_value_subset:.4f}")
if p_value_subset < alpha:
    print("Reject H₀: Significant difference even with smaller sample.")
else:
    print("Fail to reject H₀: Insufficient evidence with small sample.")
```

## Common Pitfalls to Avoid

- **P-Hacking**: Testing until a low p-value is found distorts truth—predefine hypotheses before analysis.
- **Small Samples**: With 10 rows, results are shaky—use the full dataset (1,338 rows here) for reliability.
- **Misinterpretation**: A p-value isn’t the probability H₀ is true; it’s the chance of data under H₀—focus on evidence, not proof.

Let’s check data normality to justify the t-test:

```python
# Shapiro-Wilk test for normality (p > 0.05 suggests normality)
from scipy.stats import shapiro
shapiro_smokers, shapiro_non_smokers = shapiro(smokers), shapiro(non_smokers)
print(f"Smokers Normality - p-value: {shapiro_smokers.pvalue:.4f}")
print(f"Non-Smokers Normality - p-value: {shapiro_non_smokers.pvalue:.4f}")
if shapiro_smokers.pvalue > 0.05 and shapiro_non_smokers.pvalue > 0.05:
    print("Both groups are approximately normal, supporting t-test use.")
else:
    print("Normality assumption may be violated; consider non-parametric tests.")
```
