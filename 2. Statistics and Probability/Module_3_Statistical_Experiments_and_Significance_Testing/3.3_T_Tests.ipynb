# 3.3 T-Tests
Imagine comparing the average scores of two soccer teams after a season, or tasting two coffee blends to see which one’s stronger—that’s what t-tests are for! They help you figure out if the difference between two group means is real or just a fluke. It’s like being a referee or a coffee connoisseur, making a fair call based on the numbers!

## What Are T-Tests?
A t-test compares the means of two groups to see if they differ significantly. It’s a statistical tool to decide if the difference is due to chance or a real effect. There are two main types:

- **Independent t-test**: Compares two separate groups (e.g., team A vs. team B scores).
- **Paired t-test**: Compares the same group before and after an intervention (e.g., coffee taste before vs. after roasting).

We’ll generate custom data to compare satisfaction scores for two versions of a product. Let’s create and analyze it with Python:

````python
import numpy as np
import scipy.stats as stats

# Generate custom dataset: satisfaction scores for two product versions
np.random.seed(42)
version_a = np.random.normal(loc=75, scale=10, size=30)  # Version A scores (mean 75, std 10)
version_b = np.random.normal(loc=80, scale=10, size=30)  # Version B scores (mean 80, std 10)

# Perform independent t-test (assuming unequal variances)
t_stat, p_value = stats.ttest_ind(version_a, version_b, equal_var=False)

# Print results with means and sample sizes
print(f"Mean - Version A: {version_a.mean():.2f}, Version B: {version_b.mean():.2f}")
print(f"Sample Size - Version A: {len(version_a)}, Version B: {len(version_b)}")
print(f"T-Statistic: {t_stat:.2f}")
print(f"P-Value: {p_value:.4f}")

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H₀: There is a significant difference in satisfaction between versions.")
else:
    print("Fail to reject H₀: No significant difference in satisfaction.")
````

This code generates 30 satisfaction scores for each version (means around 75 and 80), runs an independent t-test, and yields a p-value (e.g., 0.0218), suggesting a significant difference in satisfaction.

## Why Is This Necessary?

## In-Depth Explanation with Related Concepts

### How It Works
A t-test calculates a t-statistic based on the difference between means, adjusted for sample size and variability. The p-value comes from the t-distribution, showing how rare the result is under H₀. For independent t-tests, groups are assumed independent; for paired t-tests, differences are analyzed within pairs.

### Key Concepts

- **Degrees of Freedom (df)**: Reflects sample size (e.g., 30 + 30 - 2 = 58 for this test), affecting the t-distribution shape.
- **Equal Variance Assumption**: The t-test assumes equal variances unless specified otherwise (we used `equal_var=False`).
- **Effect Size**: Measures the magnitude of the difference—let’s compute Cohen’s d:

```python
# Calculate Cohen’s d for effect size
mean_diff = version_b.mean() - version_a.mean()
pooled_std = np.sqrt((version_a.var(ddof=1) * (len(version_a) - 1) + version_b.var(ddof=1) * (len(version_b) - 1)) / (len(version_a) + len(version_b) - 2))
cohens_d = mean_diff / pooled_std
print(f"Cohen’s d (Effect Size): {cohens_d:.2f}")
```

- **Paired vs. Independent**: Paired t-tests reduce variability by focusing on within-subject differences.

A moderate effect size (e.g., 0.5) with significance suggests a noticeable but not overwhelming difference.

### Intuitive Analogy
Think of it as a taste test duel. The t-statistic is the score gap, and the p-value is the crowd’s surprise—if it’s low, the winner’s lead is real, not a fluke!

## Practical Insights

- **Sample Size**: 30 per group is a good start; larger samples (e.g., 50+) refine p-values—adjust as needed.
- **Variance**: Unequal variances (`equal_var=False`) suit real-world data with different spreads—check with Levene’s test if unsure:

```python
# Levene’s test for equal variances
levene_stat, levene_p = stats.levene(version_a, version_b)
print(f"Levene’s Test p-value: {levene_p:.4f}")
if levene_p < 0.05:
    print("Reject equal variance assumption, `equal_var=False` is appropriate.")
```

- **Effect Size**: A significant p-value doesn’t mean a big difference—compare means (e.g., 75 vs. 80).

## Common Pitfalls to Avoid

- **Small Samples**: With 5 scores per group, results are unreliable—use 20+ for stability.
- **Assuming Equal Variance**: Forcing equal variance with unequal data skews p-values—verify with Levene’s test.
- **Over-Interpretation**: Significance doesn’t imply causation—consider external factors (e.g., marketing).

Let’s check normality to support the t-test:

````python
# Shapiro-Wilk test for normality (p > 0.05 suggests normality)
from scipy.stats import shapiro
shapiro_a, shapiro_b = shapiro(version_a), shapiro(version_b)
print(f"Version A Normality - p-value: {shapiro_a.pvalue:.4f}")
print(f"Version B Normality - p-value: {shapiro_b.pvalue:.4f}")
if shapiro_a.pvalue > 0.05 and shapiro_b.pvalue > 0.05:
    print("Both groups are approximately normal, supporting t-test use.")
else:
    print("Normality assumption may be violated; consider non-parametric tests.")
````
