# 01. Convexity & Convex Optimization

Welcome to **Optimization Theory for AI/ML**!

In Machine Learning, training an algorithm is fundamentally an **optimization problem**: finding the parameter weights $\mathbf{w}^*$ that minimize a loss function $L(\mathbf{w})$.
Understanding **Convexity** is the holy grail of optimization because convex problems are guaranteed to have **no bad local minima or saddle points**—every local minimum is automatically the **global minimum**.

---

## 1. Convex Sets

### Definition:
A set $\mathcal{C} \subseteq \mathbb{R}^n$ is **convex** if the line segment connecting any two points $\mathbf{x}, \mathbf{y} \in \mathcal{C}$ lies entirely inside $\mathcal{C}$:

$$\theta \mathbf{x} + (1 - \theta)\mathbf{y} \in \mathcal{C} \quad \text{for all } \mathbf{x}, \mathbf{y} \in \mathcal{C} \text{ and } \theta \in [0, 1]$$

- **Convex Sets**: Circles, solid spheres, half-spaces, hyperplanes, polyhedra, positive semi-definite cone $\mathcal{S}_+^n$.
- **Non-Convex Sets**: Donuts (annulus), star shapes, disjoint clusters.

## 2. Convex Functions

### Definition:
A function $f: \mathcal{C} \to \mathbb{R}$ defined on a convex set $\mathcal{C}$ is **convex** if for all $\mathbf{x}, \mathbf{y} \in \mathcal{C}$ and $\theta \in [0, 1]$:

$$f(\theta \mathbf{x} + (1 - \theta)\mathbf{y}) \leq \theta f(\mathbf{x}) + (1 - \theta) f(\mathbf{y})$$

- **Geometric Meaning**: The secant line connecting $(x, f(x))$ and $(y, f(y))$ always lies **above or on** the graph of $f$.
- **Strictly Convex**: If the inequality is strict ($<$) for all $\mathbf{x} \neq \mathbf{y}$ and $\theta \in (0, 1)$, there is at most **one unique global minimum**.

### Examples:
- **Convex**: $f(x) = x^2$, $f(x) = e^x$, $f(x) = -\log(x)$ (for $x > 0$), $f(\mathbf{x}) = \|\mathbf{x}\|_p$ (any vector norm).
- **Non-Convex**: $f(x) = \sin(x)$, $f(x) = x^3$, deep neural network loss functions.


```python
# Visualizing a Convex Function and its Secant Line
x = np.linspace(-2, 3, 200)
f = x**2

# Choose two points x1 = -1, x2 = 2
x1, x2 = -1.0, 2.0
y1, y2 = x1**2, x2**2

# Secant line between x1 and x2
theta = np.linspace(0, 1, 100)
secant_x = theta * x1 + (1 - theta) * x2
secant_y = theta * y1 + (1 - theta) * y2

plt.figure(figsize=(8, 5))
plt.plot(x, f, 'b-', linewidth=2.5, label=r'Function $f(x) = x^2$')
plt.plot(secant_x, secant_y, 'r--', linewidth=2, label='Secant Line (Always Above Curve)')
plt.scatter([x1, x2], [y1, y2], color='red', s=50)
plt.title(r"Definition of Convex Function: $f(\theta x + (1-\theta)y) \leq \theta f(x) + (1-\theta)f(y)$")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Mathematical Conditions for Convexity

### 3.1 First-Order Condition (Tangent Hyperplane Underestimator)
If $f$ is continuously differentiable:
$$f(\mathbf{y}) \geq f(\mathbf{x}) + \nabla f(\mathbf{x})^T (\mathbf{y} - \mathbf{x}) \quad \forall \mathbf{x}, \mathbf{y}$$
*Intuition: The linear first-order Taylor approximation always provides a global lower bound on $f$.*

### 3.2 Second-Order Condition (Positive Semi-Definite Hessian)
If $f$ is twice continuously differentiable:
$$\nabla^2 f(\mathbf{x}) \succeq 0 \quad (\text{Hessian } \mathbf{H} \text{ is Positive Semi-Definite for all } \mathbf{x})$$
*Intuition: The curvature is non-negative everywhere (bowl-shaped in all directions).*

## 4. Jensen's Inequality

For any convex function $f$ and random variable $X$:
$$f(\mathbb{E}[X]) \leq \mathbb{E}[f(X)]$$

### Applications in AI/ML:
- Deriving the **Evidence Lower Bound (ELBO)** in Variational Autoencoders (VAEs).
- Proving that KL Divergence is always non-negative ($D_{KL} \geq 0$).
- Information theory bounds on entropy and mutual information.


```python
# Verifying Jensen's Inequality numerically for f(x) = x^2
np.random.seed(42)
X = np.random.uniform(1, 10, size=10000)

E_X = np.mean(X)
f_E_X = E_X**2           # f(E[X])
E_f_X = np.mean(X**2)    # E[f(X)]

print(f"E[X] = {E_X:.4f}")
print(f"f(E[X]) = (E[X])^2 = {f_E_X:.4f}")
print(f"E[f(X)] = E[X^2]   = {E_f_X:.4f}")
print(f"Is f(E[X]) <= E[f(X)]? {f_E_X <= E_f_X} (Difference: {E_f_X - f_E_X:.4f})")
```

---

## 5. The Fundamental Theorem of Convex Optimization

### Theorem:
Let $f: \mathcal{C} \to \mathbb{R}$ be a convex function defined on a convex set $\mathcal{C}$.
**Any local minimum of $f$ is also a global minimum!**
If $f$ is strictly convex, the global minimum is **unique**.

### Why Convex Models are Beloved:
Linear Regression, Ridge/Lasso Regression, Logistic Regression, and Support Vector Machines (SVMs) are all strictly convex problems. Their optimization algorithms (e.g. Newton-Raphson, coordinate descent, convex solvers) are guaranteed to find the true optimal parameters!


---

## 6. Summary & Key Takeaways

1. **Convex sets** contain all connecting line segments; **convex functions** lie below all secant lines.
2. A function is convex iff its **Hessian matrix is Positive Semi-Definite ($\nabla^2 f \succeq 0$)**.
3. **Jensen's Inequality** $f(\mathbb{E}[X]) \leq \mathbb{E}[f(X)]$ is the mathematical bedrock of Variational Inference (VAEs).
4. In convex optimization, **local optimality $\implies$ global optimality**.
