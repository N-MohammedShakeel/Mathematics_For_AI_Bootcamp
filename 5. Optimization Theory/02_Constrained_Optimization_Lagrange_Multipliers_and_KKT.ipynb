# 02. Constrained Optimization (Lagrange Multipliers & KKT)

**How mathematical constraints shape machine learning: The method of Lagrange Multipliers, Karush-Kuhn-Tucker (KKT) conditions, and Support Vector Machines (SVMs).**

---

## 1. What is Constrained Optimization?

In unconstrained optimization, we find $\min f(\mathbf{x})$ freely across all $\mathbf{x} \in \mathbb{R}^n$.
In **Constrained Optimization**, our solution must satisfy physical, budget, probability, or geometric boundaries:

$$\min_{\mathbf{x}} f(\mathbf{x}) \quad \text{subject to} \quad \begin{cases} g_i(\mathbf{x}) \leq 0 & (i = 1, \dots, m) \quad \text{[Inequality constraints]} \\ h_j(\mathbf{x}) = 0 & (j = 1, \dots, p) \quad \text{[Equality constraints]} \end{cases}$$

---

## 2. Equality Constraints: The Lagrange Multiplier Method

Suppose we want to minimize $f(x, y)$ subject to $g(x, y) = 0$.

### Geometric Intuition:
At the constrained optimum $\mathbf{x}^*$, the contour line of the objective $f$ must be **tangent** to the constraint curve $g=0$.
Therefore, their gradient vectors must point in the exact same (or opposite) direction:
$$\nabla f(\mathbf{x}^*) = -\lambda \nabla g(\mathbf{x}^*)$$

### The Lagrangian Function:
$$\mathcal{L}(\mathbf{x}, \lambda) = f(\mathbf{x}) + \lambda g(\mathbf{x})$$

Setting the gradient $\nabla_{\mathbf{x}, \lambda} \mathcal{L} = \mathbf{0}$ yields the optimal solution!


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Example: Minimize f(x, y) = x^2 + y^2 subject to x + y = 2
# Objective function
def objective(v):
    x, y = v
    return x**2 + y**2

# Equality constraint: x + y - 2 = 0
constraints = ({'type': 'eq', 'fun': lambda v: v[0] + v[1] - 2.0})

res = minimize(objective, [0.0, 0.0], constraints=constraints)
print(f"Optimal point (x, y): {res.x}")
print(f"Minimum value f(x, y): {res.fun:.4f}")
print("Analytical solution: x=1, y=1, f(1,1)=2. Matches perfectly!")
```

## 3. Inequality Constraints: Karush-Kuhn-Tucker (KKT) Conditions

When we have inequality constraints $g_i(\mathbf{x}) \leq 0$, the Lagrangian is:
$$\mathcal{L}(\mathbf{x}, \mathbf{\lambda}, \mathbf{\nu}) = f(\mathbf{x}) + \sum_{i=1}^m \lambda_i g_i(\mathbf{x}) + \sum_{j=1}^p \nu_j h_j(\mathbf{x})$$

### The 4 Fundamental KKT Conditions for Optimality:
1. **Stationarity**:
   $$\nabla_{\mathbf{x}} f(\mathbf{x}^*) + \sum_{i=1}^m \lambda_i^* \nabla g_i(\mathbf{x}^*) + \sum_{j=1}^p \nu_j^* \nabla h_j(\mathbf{x}^*) = \mathbf{0}$$
2. **Primal Feasibility**:
   $$g_i(\mathbf{x}^*) \leq 0 \quad \text{and} \quad h_j(\mathbf{x}^*) = 0$$
3. **Dual Feasibility**:
   $$\lambda_i^* \geq 0 \quad (\text{Lagrange multipliers for inequalities must be non-negative})$$
4. **Complementary Slackness**:
   $$\lambda_i^* g_i(\mathbf{x}^*) = 0 \quad \forall i$$
   - Either the constraint is **active / binding** ($g_i(\mathbf{x}^*) = 0$, $\lambda_i^* > 0$)
   - Or the constraint is **inactive** ($g_i(\mathbf{x}^*) < 0$, $\lambda_i^* = 0$)


```python
# Visualizing Constrained Optimization Contours
x = np.linspace(-1, 3, 200)
y = np.linspace(-1, 3, 200)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2 # Objective: f(x,y) = x^2 + y^2

plt.figure(figsize=(8, 7))
contours = plt.contour(X, Y, Z, levels=[0.5, 1, 2, 4, 8, 12], cmap='viridis')
plt.clabel(contours, inline=True, fontsize=9)

# Constraint line: x + y = 2 -> y = 2 - x
plt.plot(x, 2 - x, 'r-', linewidth=2.5, label='Constraint: $x + y = 2$')
plt.scatter([1.0], [1.0], color='red', s=100, zorder=5, label='Optimal Point (1, 1), f=2')

# Draw gradient vectors at optimal point
plt.arrow(1.0, 1.0, 0.4, 0.4, head_width=0.08, color='blue', label=r'$\nabla f$')
plt.arrow(1.0, 1.0, -0.4, -0.4, head_width=0.08, color='purple', label=r'$-\lambda \nabla g$')

plt.title("Lagrange Multipliers: Tangency of Objective Contours and Constraint")
plt.xlabel("x"); plt.ylabel("y")
plt.xlim(-0.5, 2.5); plt.ylim(-0.5, 2.5)
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Real AI/ML Application: Support Vector Machines (SVM)

In Support Vector Machines, we want to find a maximum margin hyperplane $\mathbf{w}^T \mathbf{x} + b = 0$ that correctly separates two classes:
$$\min_{\mathbf{w}, b} \frac{1}{2} \|\mathbf{w}\|_2^2 \quad \text{subject to} \quad 1 - y_i(\mathbf{w}^T \mathbf{x}_i + b) \leq 0 \quad (i=1, \dots, N)$$

Using the KKT Complementary Slackness condition $\alpha_i [1 - y_i(\mathbf{w}^T \mathbf{x}_i + b)] = 0$:
- For points far from the boundary: $1 - y_i(\mathbf{w}^T \mathbf{x}_i + b) < 0 \implies \alpha_i = 0$ (they have **no effect** on the decision boundary).
- For points lying exactly on the margin boundary: $\alpha_i > 0$. These special points are the **Support Vectors**!


---

## 5. Summary & Key Takeaways

1. **Lagrange Multipliers** turn constrained optimization problems into unconstrained Lagrangian functions $\mathcal{L}(\mathbf{x}, \lambda)$.
2. **KKT Conditions** generalize Lagrange multipliers to inequality constraints via complementary slackness $\lambda_i g_i(\mathbf{x}) = 0$.
3. In **SVMs**, complementary slackness explains why only the support vectors determine the separating hyperplane.
